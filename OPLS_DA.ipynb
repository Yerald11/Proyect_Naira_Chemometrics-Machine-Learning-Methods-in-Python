```python
# ===========================================================================
# OPLS-DA INTERACTIVO - SOLO ENTRENAMIENTO / CALIBRACIÓN
# Flujo:
# CSV -> SNV -> Savitzky-Golay -> OPLS-DA preliminar
# -> diagnóstico -> eliminación manual -> OPLS-DA final
# ===========================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from scipy.stats import f
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

plt.style.use("default")


# ===========================================================================
# FUNCIONES DE ENTRADA
# ===========================================================================

def pedir_texto(mensaje, default=None):
    if default is None:
        return input(mensaje).strip()
    txt = input(f"{mensaje} [{default}]: ").strip()
    return default if txt == "" else txt


def pedir_entero(mensaje, default=None, minimo=None):
    while True:
        txt = pedir_texto(mensaje, default)
        try:
            valor = int(txt)
            if minimo is not None and valor < minimo:
                print(f"Debe ser >= {minimo}")
                continue
            return valor
        except ValueError:
            print("Ingresa un número entero válido.")


def pedir_si_no(mensaje, default="s"):
    while True:
        txt = pedir_texto(f"{mensaje} (s/n)", default).lower()
        if txt in ["s", "si", "sí", "y", "yes"]:
            return True
        if txt in ["n", "no"]:
            return False
        print("Responde con s o n.")


def pedir_confianza():
    while True:
        txt = pedir_texto("Nivel de confianza para diagnóstico T²: escribe 95 o 99", "95")
        if txt in ["95", "0.95"]:
            return 0.95
        if txt in ["99", "0.99"]:
            return 0.99
        print("Solo se permite 95 o 99.")


# ===========================================================================
# PREPROCESAMIENTO
# ===========================================================================

def listar_csv(carpeta):
    archivos = [f for f in os.listdir(carpeta) if f.lower().endswith(".csv")]
    archivos.sort()
    return archivos


def aplicar_snv(X, ddof=0):
    media = X.mean(axis=1, keepdims=True)
    std = X.std(axis=1, ddof=ddof, keepdims=True)

    if np.any(std == 0):
        raise ValueError("Hay espectros con desviación estándar cero. No se puede aplicar SNV.")

    return (X - media) / std


def aplicar_savgol(X, longitudes_onda, window_length, polyorder, deriv):
    if window_length % 2 == 0:
        raise ValueError("La ventana Savitzky-Golay debe ser impar.")

    if window_length <= polyorder:
        raise ValueError("La ventana debe ser mayor que el orden del polinomio.")

    diffs = np.diff(longitudes_onda)

    if not np.allclose(diffs, diffs[0]):
        print("[ADVERTENCIA] Longitudes de onda no perfectamente equiespaciadas. Se usará delta promedio.")

    delta = np.mean(diffs)

    X_sg = savgol_filter(
        X,
        window_length=window_length,
        polyorder=polyorder,
        deriv=deriv,
        delta=delta,
        axis=1,
        mode="interp"
    )

    return X_sg, delta


# ===========================================================================
# OPLS-DA BINARIO Y MULTICLASE ONE-VS-REST
# ===========================================================================

def ajustar_opls_binario(X, y_bin, n_orthogonal=1):
    """
    OPLS práctico para una respuesta binaria 0/1.
    Extrae 1 componente predictiva y n componentes ortogonales.
    """
    X = np.asarray(X, dtype=float)
    y = np.asarray(y_bin, dtype=float).reshape(-1, 1)

    n, p = X.shape

    X_mean = X.mean(axis=0)
    y_mean = y.mean(axis=0)

    Xc = X - X_mean
    yc = y - y_mean

    # Componente predictiva, correlacionada con Y
    w_pred = Xc.T @ yc
    w_pred = w_pred.ravel()

    norm_w = np.linalg.norm(w_pred)
    if norm_w == 0:
        raise ValueError("No se pudo calcular componente predictiva: Y no tiene variación útil.")

    w_pred = w_pred / norm_w
    t_pred = Xc @ w_pred

    denom_t = t_pred.T @ t_pred
    if denom_t == 0:
        raise ValueError("Score predictivo degenerado.")

    p_pred = (Xc.T @ t_pred) / denom_t
    c_pred = float((yc.ravel().T @ t_pred) / denom_t)

    X_res = Xc - np.outer(t_pred, p_pred)

    W_orth = []
    P_orth = []
    T_orth = []

    # Componentes ortogonales: variación de X no relacionada con Y
    for _ in range(n_orthogonal):
        # Dirección ortogonal construida a partir del loading predictivo
        w_orth = p_pred - w_pred * np.dot(w_pred, p_pred)

        norm_wo = np.linalg.norm(w_orth)
        if norm_wo < 1e-12:
            break

        w_orth = w_orth / norm_wo
        t_orth = X_res @ w_orth

        denom_to = t_orth.T @ t_orth
        if denom_to < 1e-12:
            break

        p_orth = (X_res.T @ t_orth) / denom_to

        X_res = X_res - np.outer(t_orth, p_orth)

        W_orth.append(w_orth)
        P_orth.append(p_orth)
        T_orth.append(t_orth)

    if len(W_orth) > 0:
        W_orth = np.vstack(W_orth)
        P_orth = np.vstack(P_orth)
        T_orth = np.vstack(T_orth).T
    else:
        W_orth = np.empty((0, p))
        P_orth = np.empty((0, p))
        T_orth = np.empty((n, 0))

    y_hat = y_mean[0] + c_pred * t_pred

    scores_all = np.column_stack([t_pred, T_orth])
    var_scores = np.var(scores_all, axis=0, ddof=1)
    var_scores = np.where(var_scores == 0, np.nan, var_scores)

    QX = np.sum(X_res ** 2, axis=1)
    EY = np.abs(y.ravel() - y_hat)

    return {
        "X_mean": X_mean,
        "y_mean": y_mean[0],
        "w_pred": w_pred,
        "p_pred": p_pred,
        "c_pred": c_pred,
        "W_orth": W_orth,
        "P_orth": P_orth,
        "t_pred": t_pred,
        "T_orth": T_orth,
        "scores_all": scores_all,
        "var_scores": var_scores,
        "QX_cal": QX,
        "EY_cal": EY,
        "y_hat_cal": y_hat,
        "n_orthogonal_real": W_orth.shape[0]
    }


def predecir_opls_binario(X, modelo):
    X = np.asarray(X, dtype=float)
    Xc = X - modelo["X_mean"]

    X_def = Xc.copy()
    T_orth_pred = []

    for w_o, p_o in zip(modelo["W_orth"], modelo["P_orth"]):
        t_o = X_def @ w_o
        X_def = X_def - np.outer(t_o, p_o)
        T_orth_pred.append(t_o)

    t_pred = X_def @ modelo["w_pred"]
    y_hat = modelo["y_mean"] + modelo["c_pred"] * t_pred

    X_hat = np.outer(t_pred, modelo["p_pred"])

    if len(T_orth_pred) > 0:
        T_orth_pred = np.vstack(T_orth_pred).T
        for j in range(T_orth_pred.shape[1]):
            X_hat += np.outer(T_orth_pred[:, j], modelo["P_orth"][j])
    else:
        T_orth_pred = np.empty((X.shape[0], 0))

    residual = Xc - X_hat
    QX = np.sum(residual ** 2, axis=1)

    scores_all = np.column_stack([t_pred, T_orth_pred])

    return y_hat, t_pred, T_orth_pred, scores_all, QX


def entrenar_opls_da_multiclase(X, y_labels, n_orthogonal=1, alpha=0.95):
    clases = np.unique(y_labels)
    modelos = {}
    respuestas = np.zeros((X.shape[0], len(clases)))
    QX_por_modelo = np.zeros((X.shape[0], len(clases)))
    T2_por_modelo = np.zeros((X.shape[0], len(clases)))
    T2_limites = {}

    for j, clase in enumerate(clases):
        y_bin = (y_labels == clase).astype(float)

        modelo = ajustar_opls_binario(X, y_bin, n_orthogonal=n_orthogonal)

        y_hat, t_pred, T_orth, scores_all, QX = predecir_opls_binario(X, modelo)

        a = scores_all.shape[1]
        n = X.shape[0]

        T2 = np.nansum((scores_all ** 2) / modelo["var_scores"], axis=1)

        if n > a:
            T2_lim = a * (n - 1) / (n - a) * f.ppf(alpha, a, n - a)
        else:
            T2_lim = np.nan

        modelos[clase] = modelo
        respuestas[:, j] = y_hat
        QX_por_modelo[:, j] = QX
        T2_por_modelo[:, j] = T2
        T2_limites[clase] = T2_lim

    y_pred = clases[np.argmax(respuestas, axis=1)]

    return {
        "clases": clases,
        "modelos": modelos,
        "respuestas": respuestas,
        "y_pred": y_pred,
        "QX_por_modelo": QX_por_modelo,
        "T2_por_modelo": T2_por_modelo,
        "T2_limites": T2_limites
    }


def diagnostico_oplsda_clase_real(oplsda_res, y_labels):
    clases = oplsda_res["clases"]
    respuestas = oplsda_res["respuestas"]
    QX_m = oplsda_res["QX_por_modelo"]
    T2_m = oplsda_res["T2_por_modelo"]

    T2_real = np.zeros(len(y_labels))
    QX_real = np.zeros(len(y_labels))
    EY_real = np.zeros(len(y_labels))
    respuesta_real = np.zeros(len(y_labels))
    T2_lim_real = np.zeros(len(y_labels))

    for i, clase_real in enumerate(y_labels):
        j = np.where(clases == clase_real)[0][0]
        T2_real[i] = T2_m[i, j]
        QX_real[i] = QX_m[i, j]
        respuesta_real[i] = respuestas[i, j]
        EY_real[i] = abs(1.0 - respuestas[i, j])
        T2_lim_real[i] = oplsda_res["T2_limites"][clase_real]

    return T2_real, QX_real, EY_real, respuesta_real, T2_lim_real


# ===========================================================================
# GRAFICOS
# ===========================================================================

def graficar_oplsda_modelos(oplsda_res, y_labels, titulo="preliminar"):
    clases = oplsda_res["clases"]
    modelos = oplsda_res["modelos"]

    for clase_modelo in clases:
        modelo = modelos[clase_modelo]
        t_pred = modelo["t_pred"]

        if modelo["T_orth"].shape[1] > 0:
            t_orth1 = modelo["T_orth"][:, 0]
            ylabel = "t_orthogonal 1"
        else:
            t_orth1 = np.zeros_like(t_pred)
            ylabel = "0"

        plt.figure(figsize=(8, 6))

        for clase in np.unique(y_labels):
            idx = y_labels == clase
            plt.scatter(
                t_pred[idx],
                t_orth1[idx],
                label=clase,
                s=65,
                edgecolors="black",
                linewidths=0.6,
                alpha=0.85
            )

        for i in range(len(y_labels)):
            plt.text(t_pred[i], t_orth1[i], str(i), fontsize=8, alpha=0.65)

        plt.axhline(0, color="black", linestyle=":", alpha=0.5)
        plt.axvline(0, color="black", linestyle=":", alpha=0.5)
        plt.xlabel("t_predictivo")
        plt.ylabel(ylabel)
        plt.title(f"OPLS-DA {titulo} - Modelo {clase_modelo} vs resto")
        plt.legend()
        plt.grid(alpha=0.2)
        plt.tight_layout()
        plt.show()


def graficar_diagnostico_oplsda(T2, QX, EY, T2_lim_real, y_labels, titulo="preliminar"):
    fig, ax = plt.subplots(1, 3, figsize=(18, 5))

    colores_t2 = np.where(T2 > T2_lim_real, "crimson", "teal")

    ax[0].bar(range(len(T2)), T2, color=colores_t2, alpha=0.8)
    ax[0].plot(T2_lim_real, color="red", linestyle="--", label="Límite T² por clase")
    ax[0].set_title(f"Hotelling T² clase real - {titulo}")
    ax[0].set_xlabel("Índice")
    ax[0].set_ylabel("T²")
    ax[0].legend()
    ax[0].grid(axis="y", alpha=0.25)

    ax[1].bar(range(len(QX)), QX, color="slateblue", alpha=0.8)
    ax[1].set_title(f"Q residual X clase real - {titulo}")
    ax[1].set_xlabel("Índice")
    ax[1].set_ylabel("QX")
    ax[1].grid(axis="y", alpha=0.25)

    ax[2].bar(range(len(EY)), EY, color="darkorange", alpha=0.8)
    ax[2].set_title(f"Error Y clase real - {titulo}")
    ax[2].set_xlabel("Índice")
    ax[2].set_ylabel("|1 - respuesta clase real|")
    ax[2].grid(axis="y", alpha=0.25)

    plt.tight_layout()
    plt.show()


def graficar_respuestas_oplsda(oplsda_res, titulo="final"):
    respuestas = oplsda_res["respuestas"]
    clases = oplsda_res["clases"]

    plt.figure(figsize=(11, 5))

    for j, clase in enumerate(clases):
        plt.plot(respuestas[:, j], marker="o", linestyle="", label=f"Respuesta {clase}")

    plt.axhline(0.5, color="red", linestyle="--", label="Referencia 0.5")
    plt.xlabel("Índice de muestra")
    plt.ylabel("Respuesta OPLS-DA")
    plt.title(f"Respuestas continuas OPLS-DA - {titulo}")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()


def graficar_espectros(X_raw, X_proc, longitudes_onda):
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    ax[0].plot(longitudes_onda, X_raw.T, alpha=0.45)
    ax[0].set_title("Espectros originales")
    ax[0].set_xlabel("Longitud de onda")
    ax[0].set_ylabel("Absorbancia")
    ax[0].grid(alpha=0.2)

    ax[1].plot(longitudes_onda, X_proc.T, alpha=0.45)
    ax[1].set_title("Espectros preprocesados")
    ax[1].set_xlabel("Longitud de onda")
    ax[1].set_ylabel("Intensidad preprocesada")
    ax[1].grid(alpha=0.2)

    plt.tight_layout()
    plt.show()


# ===========================================================================
# 1. CONFIGURACION INTERACTIVA
# ===========================================================================

print("=" * 75)
print("OPLS-DA INTERACTIVO - SOLO ENTRENAMIENTO / CALIBRACIÓN")
print("Implementación multiclase one-vs-rest")
print("=" * 75)

carpeta = pedir_texto("\nIngresa la carpeta donde están los CSV", ".")
carpeta = os.path.abspath(carpeta)

if not os.path.isdir(carpeta):
    raise FileNotFoundError(f"No existe la carpeta: {carpeta}")

csv_disponibles = listar_csv(carpeta)

if len(csv_disponibles) == 0:
    raise FileNotFoundError("No se encontraron archivos CSV.")

print("\nCSV encontrados:")
for i, archivo in enumerate(csv_disponibles, start=1):
    print(f"{i}. {archivo}")

seleccion_txt = pedir_texto("\nEscribe los números de los CSV separados por coma. Ejemplo: 1,2,3")
indices_csv = [int(x.strip()) - 1 for x in seleccion_txt.split(",")]
archivos_seleccionados = [csv_disponibles[i] for i in indices_csv]

etiquetas_clase = {}

print("\nAsigna una etiqueta/clase para cada archivo.")
for archivo in archivos_seleccionados:
    etiqueta_default = os.path.splitext(archivo)[0]
    etiqueta = pedir_texto(f"Etiqueta para {archivo}", etiqueta_default)
    etiquetas_clase[archivo] = etiqueta

sg_window = pedir_entero("\nVentana Savitzky-Golay", default=17, minimo=3)
sg_polyorder = pedir_entero("Orden del polinomio Savitzky-Golay", default=2, minimo=1)
sg_deriv = pedir_entero("Orden de derivada Savitzky-Golay", default=1, minimo=0)

n_orthogonal = pedir_entero("\nNúmero de componentes ortogonales OPLS", default=1, minimo=0)
alpha = pedir_confianza()

exportar = pedir_si_no("\n¿Deseas exportar resultados CSV?", default="s")


# ===========================================================================
# 2. CARGA DE DATOS
# ===========================================================================

dfs = []
y_labels = []

for archivo in archivos_seleccionados:
    ruta = os.path.join(carpeta, archivo)
    df = pd.read_csv(ruta)
    clase = etiquetas_clase[archivo]

    dfs.append(df)
    y_labels.extend([clase] * len(df))

    print(f"[OK] {archivo}: {df.shape[0]} muestras, clase = {clase}")

df_global = pd.concat(dfs, ignore_index=True)
y_labels = np.array(y_labels)

columnas_convertidas = pd.to_numeric(df_global.columns, errors="coerce")
columnas_espectrales = df_global.columns[columnas_convertidas.notna()]
columnas_metadata = df_global.columns[columnas_convertidas.isna()]

longitudes_onda = columnas_convertidas[columnas_convertidas.notna()].astype(float).values
X_raw = df_global[columnas_espectrales].astype(float).values

orden = np.argsort(longitudes_onda)
longitudes_onda = longitudes_onda[orden]
X_raw = X_raw[:, orden]

print("\nColumnas de metadatos omitidas:")
print(list(columnas_metadata))

print(f"\nMatriz espectral cruda: {X_raw.shape} = muestras x variables")

if np.isnan(X_raw).any():
    raise ValueError("La matriz contiene NaN.")

if np.isinf(X_raw).any():
    raise ValueError("La matriz contiene infinitos.")


# ===========================================================================
# 3. PREPROCESAMIENTO
# ===========================================================================

X_snv = aplicar_snv(X_raw, ddof=0)

X_proc, delta_sg = aplicar_savgol(
    X_snv,
    longitudes_onda,
    window_length=sg_window,
    polyorder=sg_polyorder,
    deriv=sg_deriv
)

print("\nPreprocesamiento aplicado:")
print("- SNV muestra por muestra, ddof=0")
print(f"- Savitzky-Golay: ventana={sg_window}, polinomio={sg_polyorder}, derivada={sg_deriv}, delta={delta_sg:.6f}")


# ===========================================================================
# 4. OPLS-DA PRELIMINAR
# ===========================================================================

opls_pre = entrenar_opls_da_multiclase(
    X_proc,
    y_labels,
    n_orthogonal=n_orthogonal,
    alpha=alpha
)

T2_pre, QX_pre, EY_pre, respuesta_real_pre, T2_lim_pre = diagnostico_oplsda_clase_real(
    opls_pre,
    y_labels
)

print("\n" + "=" * 75)
print("OPLS-DA PRELIMINAR - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels, opls_pre["y_pred"]))

graficar_espectros(X_raw, X_proc, longitudes_onda)
graficar_oplsda_modelos(opls_pre, y_labels, titulo="preliminar")
graficar_respuestas_oplsda(opls_pre, titulo="preliminar")
graficar_diagnostico_oplsda(T2_pre, QX_pre, EY_pre, T2_lim_pre, y_labels, titulo="preliminar")

cm = confusion_matrix(y_labels, opls_pre["y_pred"], labels=opls_pre["clases"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=opls_pre["clases"])
disp.plot(cmap="Blues", values_format="d")
plt.title("Matriz de confusión aparente - OPLS-DA preliminar")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

df_diag_pre = pd.DataFrame({
    "Indice": np.arange(len(y_labels)),
    "Clase_real": y_labels,
    "Clase_predicha": opls_pre["y_pred"],
    "Correcta": y_labels == opls_pre["y_pred"],
    "Respuesta_clase_real": respuesta_real_pre,
    "T2_clase_real": T2_pre,
    "QX_clase_real": QX_pre,
    "Error_Y_clase_real": EY_pre,
    "T2_fuera_limite": T2_pre > T2_lim_pre
})

print("\nMuestras con mayor Hotelling T²:")
display(df_diag_pre.sort_values("T2_clase_real", ascending=False).head(15))

print("\nMuestras con mayor Q residual X:")
display(df_diag_pre.sort_values("QX_clase_real", ascending=False).head(15))

print("\nMuestras con mayor error Y:")
display(df_diag_pre.sort_values("Error_Y_clase_real", ascending=False).head(15))

print("\nMuestras mal clasificadas en calibración aparente:")
display(df_diag_pre[df_diag_pre["Correcta"] == False])


# ===========================================================================
# 5. ELIMINACIÓN MANUAL DESPUÉS DEL DIAGNÓSTICO
# ===========================================================================

eliminar = pedir_si_no(
    "\nDespués de ver los diagnósticos OPLS-DA, ¿deseas eliminar muestras manualmente?",
    default="n"
)

indices_eliminar = []

if eliminar:
    txt = pedir_texto("Escribe los índices globales a eliminar separados por coma. Ejemplo: 54 o 54,67")
    if txt.strip() != "":
        indices_eliminar = [int(x.strip()) for x in txt.split(",")]

if len(indices_eliminar) > 0:
    indices_eliminar = np.array(indices_eliminar, dtype=int)

    if np.any(indices_eliminar < 0) or np.any(indices_eliminar >= X_raw.shape[0]):
        raise ValueError("Hay índices a eliminar fuera del rango.")

    mascara = np.ones(X_raw.shape[0], dtype=bool)
    mascara[indices_eliminar] = False

    print("\nMuestras eliminadas manualmente:")
    for idx in indices_eliminar:
        print(f"- Índice global {idx}, clase {y_labels[idx]}")

    X_raw_final = X_raw[mascara]
    X_proc_final = X_proc[mascara]
    y_labels_final = y_labels[mascara]
    df_global_final = df_global.loc[mascara].reset_index(drop=True)

else:
    print("\nNo se eliminaron muestras. El OPLS-DA final usará todas las muestras.")
    X_raw_final = X_raw.copy()
    X_proc_final = X_proc.copy()
    y_labels_final = y_labels.copy()
    df_global_final = df_global.copy()

print(f"\nMatriz final para OPLS-DA: {X_proc_final.shape}")


# ===========================================================================
# 6. OPLS-DA FINAL
# ===========================================================================

opls_final = entrenar_opls_da_multiclase(
    X_proc_final,
    y_labels_final,
    n_orthogonal=n_orthogonal,
    alpha=alpha
)

T2_final, QX_final, EY_final, respuesta_real_final, T2_lim_final = diagnostico_oplsda_clase_real(
    opls_final,
    y_labels_final
)

print("\n" + "=" * 75)
print("OPLS-DA FINAL - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels_final, opls_final["y_pred"]))

graficar_espectros(X_raw_final, X_proc_final, longitudes_onda)
graficar_oplsda_modelos(opls_final, y_labels_final, titulo="final")
graficar_respuestas_oplsda(opls_final, titulo="final")
graficar_diagnostico_oplsda(T2_final, QX_final, EY_final, T2_lim_final, y_labels_final, titulo="final")

cm = confusion_matrix(y_labels_final, opls_final["y_pred"], labels=opls_final["clases"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=opls_final["clases"])
disp.plot(cmap="Blues", values_format="d")
plt.title("Matriz de confusión aparente - OPLS-DA final")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ===========================================================================
# 7. TABLAS FINALES
# ===========================================================================

df_pred_final = pd.DataFrame({
    "Indice": np.arange(len(y_labels_final)),
    "Clase_real": y_labels_final,
    "Clase_predicha_calibracion": opls_final["y_pred"],
    "Correcta": y_labels_final == opls_final["y_pred"],
    "Respuesta_clase_real": respuesta_real_final,
    "T2_clase_real": T2_final,
    "QX_clase_real": QX_final,
    "Error_Y_clase_real": EY_final,
    "T2_fuera_limite": T2_final > T2_lim_final
})

for j, clase in enumerate(opls_final["clases"]):
    df_pred_final[f"Respuesta_{clase}"] = opls_final["respuestas"][:, j]
    df_pred_final[f"T2_modelo_{clase}"] = opls_final["T2_por_modelo"][:, j]
    df_pred_final[f"QX_modelo_{clase}"] = opls_final["QX_por_modelo"][:, j]

display(df_pred_final)

print("\nMuestras con mayor T² final:")
display(df_pred_final.sort_values("T2_clase_real", ascending=False).head(15))

print("\nMuestras con mayor QX final:")
display(df_pred_final.sort_values("QX_clase_real", ascending=False).head(15))

print("\nMuestras con mayor error Y final:")
display(df_pred_final.sort_values("Error_Y_clase_real", ascending=False).head(15))


# ===========================================================================
# 8. EXPORTACIÓN
# ===========================================================================

if exportar:
    carpeta_salida = os.path.join(carpeta, "resultados_opls_da_entrenamiento")
    os.makedirs(carpeta_salida, exist_ok=True)

    df_preprocesado = pd.DataFrame(X_proc_final, columns=[str(w) for w in longitudes_onda])
    df_preprocesado.insert(0, "Clase", y_labels_final)
    df_preprocesado.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_eliminadas = pd.DataFrame({
        "Indice_original_eliminado": indices_eliminar
    })

    df_pred_final.to_csv(os.path.join(carpeta_salida, "oplsda_predicciones_calibracion_final.csv"), index=False)
    df_diag_pre.to_csv(os.path.join(carpeta_salida, "oplsda_diagnostico_preliminar.csv"), index=False)
    df_preprocesado.to_csv(os.path.join(carpeta_salida, "matriz_preprocesada_snv_sg_final.csv"), index=False)
    df_eliminadas.to_csv(os.path.join(carpeta_salida, "muestras_eliminadas.csv"), index=False)

    # Exportar parámetros por modelo/clase
    for clase, modelo in opls_final["modelos"].items():
        clase_segura = str(clase).replace(" ", "_").replace("/", "_")

        df_param = pd.DataFrame({
            "Longitud_onda": longitudes_onda,
            "w_predictivo": modelo["w_pred"],
            "p_predictivo": modelo["p_pred"]
        })

        for k in range(modelo["W_orth"].shape[0]):
            df_param[f"w_orth_{k+1}"] = modelo["W_orth"][k]
            df_param[f"p_orth_{k+1}"] = modelo["P_orth"][k]

        df_param.to_csv(
            os.path.join(carpeta_salida, f"oplsda_parametros_modelo_{clase_segura}.csv"),
            index=False
        )

    print("\n[OK] Resultados exportados en:")
    print(carpeta_salida)
```

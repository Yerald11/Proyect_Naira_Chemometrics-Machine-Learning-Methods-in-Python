```python
# ===========================================================================
# VALIDACIÓN ANIDADA POR LOTE: PLS-DA vs SVM  (Proyecto NaiRa, Hito 5/6)
#
# Flujo (idéntico para ambos modelos):
#   CSV -> SNV -> Savitzky-Golay -> (sin escalado, sin corte espectral)
#   -> CV externa agrupada por lote (estimación de desempeño)
#   -> CV interna agrupada por lote (elige VL / kernel, C, gamma)
#
# Ningún espectro de un lote usado para ajustar/elegir hiperparámetros
# aparece en el fold que se evalúa. Las réplicas de una misma unidad
# quedan siempre en el mismo grupo.
#
# Entrada interactiva (igual que tus scripts): carpeta, CSV, etiquetas
# de clase y columna de metadatos que identifica el lote.
# ===========================================================================

import os
import re
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from scipy.stats import binomtest
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import LabelBinarizer
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedGroupKFold, GridSearchCV
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, f1_score, cohen_kappa_score,
    confusion_matrix, ConfusionMatrixDisplay,
)

warnings.simplefilter("ignore", pd.errors.PerformanceWarning)

# ===========================================================================
# CONFIGURACIÓN (se exporta como audit trail)
# ===========================================================================

CARPETA_DEFAULT = r"C:\Users\YERALD"   # carpeta sugerida (como en tus scripts)

# Nomenclatura de muestra (jerárquica):  API - PRODUCTO - FABRICANTE - X - LOTE(3) - ULT(3)
#   ej. NAP-AP-BY-A-156-002
#   campo 1 (API)         nivel principal: cada API se modela por separado
#   campos 2-3            definen la CLASE dentro del API (PRODUCTO + FABRICANTE)
#   campo 5 (LOTE)        mismo número = mismo lote; otro número = otro lote de la misma clase
#   campo 4 (X) y ULT     NO se consideran
# Los campos 1-4 son alfanuméricos de longitud libre (el API, el tipo de producto y el
# laboratorio pueden variar); el separador puede ser "-" o "_".
PATRON_CODIGO = re.compile(
    r"(?P<api>[A-Za-z0-9]+)[-_](?P<prod>[A-Za-z0-9]+)[-_](?P<lab>[A-Za-z0-9]+)[-_]"
    r"(?P<campo4>[A-Za-z0-9]+)[-_](?P<lote>\d{3})[-_](?P<ult>\d{3})")
SG = dict(window_length=17, polyorder=2, deriv=1)
VL_GRID = list(range(1, 16))           # variables latentes a probar (PLS-DA)
C_GRID = [0.1, 1, 10, 100, 1000]       # SVM
GAMMA_FACTORES = [0.1, 0.3, 1, 3, 10]  # múltiplos de gamma="scale"
MAX_SPLITS_EXTERNOS = 5
SEMILLA = 0


# ===========================================================================
# MODELOS
# ===========================================================================

class PLSDA(BaseEstimator, ClassifierMixin):
    """PLS-DA con Y indicadora (one-hot), sin escalado, clase = argmax."""

    def __init__(self, n_components=3):
        self.n_components = n_components

    def fit(self, X, y):
        self.lb_ = LabelBinarizer().fit(y)
        Y = self.lb_.transform(y)
        if Y.shape[1] == 1:
            Y = np.hstack([1 - Y, Y])
        k = min(self.n_components, X.shape[0] - 1, X.shape[1])
        self.pls_ = PLSRegression(n_components=k, scale=False).fit(X, Y)
        self.classes_ = self.lb_.classes_
        return self

    def predict(self, X):
        return self.classes_[np.argmax(self.pls_.predict(X), axis=1)]


# ===========================================================================
# PREPROCESAMIENTO Y CARGA
# ===========================================================================

def snv(X):
    sd = X.std(axis=1, keepdims=True)
    if np.any(sd == 0):
        raise ValueError("Espectro con desviación estándar cero: no se puede aplicar SNV.")
    return (X - X.mean(axis=1, keepdims=True)) / sd


def pedir_texto(mensaje, default=None):
    if default is None:
        return input(f"{mensaje}: ").strip()
    txt = input(f"{mensaje} [{default}]: ").strip()
    return default if txt == "" else txt


def elegir_columna(meta, mensaje, permitir_vacio=False):
    """Pide una columna de metadatos por número o por nombre exacto."""
    for i, c in enumerate(meta, start=1):
        print(f"  {i}. {c}")
    while True:
        txt = pedir_texto(mensaje, "" if permitir_vacio else None)
        if txt == "" and permitir_vacio:
            return None
        if txt.isdigit() and 1 <= int(txt) <= len(meta):
            return meta[int(txt) - 1]
        if txt in meta:
            return txt
        print(f"Opción no válida. Escribe el número (1-{len(meta)}) o el nombre exacto.")


def parsear_codigo(texto):
    m = PATRON_CODIGO.search(str(texto))
    if not m:
        return None
    d = m.groupdict()
    d["clase"] = f"{d['api']}-{d['prod']}-{d['lab']}".upper()
    d["api"] = d["api"].upper()
    return d


def cargar():
    carpeta = os.path.abspath(pedir_texto("Carpeta donde están los CSV", CARPETA_DEFAULT))
    if not os.path.isdir(carpeta):
        raise FileNotFoundError(f"No existe la carpeta: {carpeta}")

    csvs = sorted(f for f in os.listdir(carpeta) if f.lower().endswith(".csv"))
    if not csvs:
        raise FileNotFoundError("No se encontraron archivos CSV.")

    print("\nCSV encontrados:")
    for i, a in enumerate(csvs, start=1):
        print(f"{i}. {a}")

    sel = pedir_texto("\nNúmeros de los CSV separados por coma (ej. 1,2,3)")
    idx = [int(x.strip()) - 1 for x in sel.split(",")]
    if any(i < 0 or i >= len(csvs) for i in idx):
        raise ValueError("Uno de los números está fuera de rango.")
    archivos = [csvs[i] for i in idx]

    dfs, nombres = [], []
    for a in archivos:
        d = pd.read_csv(os.path.join(carpeta, a))
        dfs.append(d)
        nombres += [a] * len(d)
    df = pd.concat(dfs, ignore_index=True).copy()
    df["__archivo"] = nombres
    df["__fila"] = df.groupby("__archivo").cumcount().values

    meta = [c for c in df.columns
            if not str(c).startswith("__") and np.isnan(pd.to_numeric(c, errors="coerce"))]

    # --- Código de muestra -> clase y lote
    origen = pedir_texto("\nDónde está el código de muestra (ej. NA2-GN-PT-A-236-002): "
                         "1 = nombre del archivo, 2 = columna de metadatos", "1")
    if origen == "2":
        print("Columnas de metadatos disponibles:")
        col_codigo = elegir_columna(meta, "Columna con el código de muestra (número o nombre)")
        codigos = df[col_codigo].astype(str)
    else:
        codigos = df["__archivo"].map(lambda a: os.path.splitext(a)[0])

    parsed = codigos.map(parsear_codigo)
    malos = sorted(set(codigos[parsed.isna()]))
    if malos:
        raise ValueError("Códigos que no cumplen API-TIPO-LAB-X-LLL-UUU: " + str(malos[:10]))
    df["__clase"] = parsed.map(lambda d: d["clase"]).values
    df["__ult"] = parsed.map(lambda d: d["ult"]).values

    if meta:
        print("\nValores distintos por archivo en cada columna de metadatos (ayuda para elegir el lote):")
        for c in meta:
            print(f"  [{c}]", df.groupby("__archivo")[c].nunique().to_dict())
    origen_lote = "1"
    if meta:
        origen_lote = pedir_texto("\nOrigen del LOTE: 1 = 3 dígitos del código (campo 5), "
                                  "2 = columna de metadatos (ej. Batch Number)", "1")
    if origen_lote == "2":
        col_lote = elegir_columna(meta, "Columna que contiene el lote (número o nombre)")
        if df[col_lote].isna().any():
            raise ValueError(f"La columna '{col_lote}' tiene valores vacíos.")
        df["__lote"] = df[col_lote].astype(str).values
    else:
        df["__lote"] = parsed.map(lambda d: d["lote"]).values

    # --- Mapa archivo -> clase -> lote (agrupa aunque los archivos estén desordenados)
    mapa = (df.groupby("__archivo")
              .agg(api=("__clase", lambda s_: s_.iloc[0].split("-")[0]),
                   clase=("__clase", "first"),
                   lote=("__lote", lambda s_: ",".join(sorted(set(s_)))),
                   espectros=("__clase", "size"))
              .reset_index().rename(columns={"__archivo": "archivo"})
              .sort_values(["api", "clase", "lote", "archivo"]).reset_index(drop=True))
    mapa["lotes_en_clase"] = mapa["clase"].map(df.groupby("__clase")["__lote"].nunique())
    print("\nAgrupación detectada a partir de los nombres de archivo (API > clase > lote):")
    for api_, sub in mapa.groupby("api"):
        print(f"\n  API {api_}  ({sub['clase'].nunique()} clases)")
        print(sub.drop(columns="api").to_string(index=False))
    if pedir_texto("\n¿La agrupación es correcta? (s/n)", "s").lower() in ("n", "no"):
        raise ValueError("Agrupación rechazada: corrige los nombres de archivo y vuelve a ejecutar.")

    # --- Clases con un solo lote: no se pueden validar con lotes no vistos
    lotes_cls = df.groupby("__clase")["__lote"].nunique()
    pocos = lotes_cls[lotes_cls < 2].index.tolist()
    col_unidad = None
    if pocos:
        print(f"\n[ADVERTENCIA] Clases con un solo lote: {pocos}")
        print("Para ellas no existe validación con lotes no vistos.")
        if meta:
            print("Columnas de metadatos (solo sirve una que identifique la unidad/tableta):")
            col_unidad = elegir_columna(
                meta, "Número o nombre de la columna de unidad (Enter = excluir esas clases)",
                permitir_vacio=True)
        if not col_unidad:
            print(f"[AVISO] Se excluyen de la validación: {pocos}")
            df = df[~df["__clase"].isin(pocos)].reset_index(drop=True)
            if df["__clase"].nunique() < 2:
                raise ValueError("Quedan menos de 2 clases tras excluir las de un solo lote.")

    num = np.array([pd.to_numeric(c, errors="coerce") for c in df.columns], dtype=float)
    es_esp = ~np.isnan(num)
    wl = num[es_esp]
    X = df.loc[:, es_esp].astype(float).values
    orden = np.argsort(wl)
    wl, X = wl[orden], X[:, orden]

    if not np.isfinite(X).all():
        raise ValueError("La matriz espectral contiene NaN o infinitos.")

    y = df["__clase"].values
    api = df["__clase"].str.split("-").str[0].values
    base = df["__clase"] + "|" + df["__lote"]                      # lote único por clase
    nivel = np.where(df["__clase"].isin(pocos), "unidad", "lote")
    if col_unidad:
        g = np.where(nivel == "unidad",
                     (base + "|" + df[col_unidad].astype(str)).values, base.values)
    else:
        g = base.values

    print("\nClases (API-PRODUCTO-FABRICANTE) y grupos de validación:")
    tabla = pd.DataFrame({
        "lotes": df.groupby("__clase")["__lote"].nunique(),
        "lista_lotes": df.groupby("__clase")["__lote"].agg(lambda s: ",".join(sorted(set(s)))),
        "espectros": df.groupby("__clase").size(),
        "nivel": pd.Series(nivel, index=y).groupby(level=0).first(),
    })
    print(tabla.to_string())

    trazas = pd.DataFrame({"archivo": df["__archivo"].values,
                           "fila_en_archivo": df["__fila"].values})
    info = {"carpeta": carpeta, "archivos": archivos, "col_unidad": col_unidad,
            "nivel": nivel, "tabla_clases": tabla, "api": api,
            "trazas": trazas, "mapa": mapa}
    return X, y, g, wl, info


# ===========================================================================
# VALIDACIÓN
# ===========================================================================

def lotes_min_por_clase(y, g):
    return int(pd.Series(g).groupby(y).nunique().min())


def validar(X, y, g, modelos, fijos):
    n_ext = min(MAX_SPLITS_EXTERNOS, lotes_min_por_clase(y, g))
    if n_ext < 2:
        raise ValueError("Se necesitan al menos 2 lotes por clase para validar por lote.")
    externo = StratifiedGroupKFold(n_splits=n_ext, shuffle=True, random_state=SEMILLA)

    pred = {nombre: np.empty(len(y), dtype=object) for nombre in modelos}
    elegidos = []

    for k, (tr, te) in enumerate(externo.split(X, y, g), start=1):
        n_int = min(3, lotes_min_por_clase(y[tr], g[tr]))
        if n_int < 2:
            # Alguna clase conserva un solo lote en entrenamiento: no se puede optimizar
            print(f"[fold {k}/{n_ext}] [AVISO] CV interna imposible (una clase queda con 1 lote "
                  f"en entrenamiento): se usan hiperparámetros FIJOS, sin optimizar.")
            for nombre, (est, grid) in modelos.items():
                m = clone(est).set_params(**fijos[nombre]).fit(X[tr], y[tr])
                pred[nombre][te] = m.predict(X[te])
                elegidos.append({"fold": k, "modelo": nombre, "optimizado": False,
                                 **{p_: str(v) for p_, v in fijos[nombre].items()}})
            continue
        interno = StratifiedGroupKFold(n_splits=n_int, shuffle=True, random_state=SEMILLA + 1)

        for nombre, (est, grid) in modelos.items():
            gs = GridSearchCV(est, grid, cv=interno, scoring="balanced_accuracy",
                              n_jobs=-1, refit=True)
            gs.fit(X[tr], y[tr], groups=g[tr])
            pred[nombre][te] = gs.predict(X[te])
            elegidos.append({"fold": k, "modelo": nombre, "optimizado": True,
                             **{p_: str(v) for p_, v in gs.best_params_.items()}})
            print(f"[fold {k}/{n_ext}] {nombre}: {gs.best_params_}")

    return pred, pd.DataFrame(elegidos), n_ext


def metricas(y, yp, g, clases, nivel):
    # A nivel de espectro
    res = {
        "Exactitud_espectros": accuracy_score(y, yp),
        "Exactitud_balanceada": balanced_accuracy_score(y, yp),
        "F1_macro": f1_score(y, yp, average="macro"),
        "Kappa": cohen_kappa_score(y, yp),
    }
    # Por grupo (voto mayoritario). El N real es el número de grupos, no de espectros.
    # Clases con >=2 lotes: grupo = lote (validación con lotes no vistos).
    # Clases con 1 lote: grupo = unidad (solo variabilidad intra-lote: optimista).
    d = pd.DataFrame({"g": g, "y": y, "yp": yp, "nivel": nivel})
    grp = d.groupby("g").agg(y=("y", "first"), yp=("yp", lambda s: s.mode().iat[0]),
                             nivel=("nivel", "first"))
    for niv, etq in [("lote", "lotes"), ("unidad", "unidades")]:
        sub = grp[grp.nivel == niv]
        if len(sub):
            ok = int((sub.y == sub.yp).sum())
            ic = binomtest(ok, len(sub)).proportion_ci(method="exact")
            res.update({f"N_{etq}": len(sub), f"Exactitud_{etq}": ok / len(sub),
                        f"IC95_{etq}_inf": ic.low, f"IC95_{etq}_sup": ic.high})
        else:
            res.update({f"N_{etq}": 0, f"Exactitud_{etq}": np.nan,
                        f"IC95_{etq}_inf": np.nan, f"IC95_{etq}_sup": np.nan})

    # Sensibilidad / especificidad por clase (espectros)
    cm = confusion_matrix(y, yp, labels=clases)
    filas = []
    for i, c in enumerate(clases):
        tp = cm[i, i]
        fn = cm[i].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - tp - fn - fp
        filas.append({"Clase": c,
                      "Sensibilidad": tp / (tp + fn) if tp + fn else np.nan,
                      "Especificidad": tn / (tn + fp) if tn + fp else np.nan})
    return res, pd.DataFrame(filas), cm


# ===========================================================================
# EJECUCIÓN
# ===========================================================================

def ejecutar(X_raw, y, g, nivel, wl, salida, extra, trazas):
    os.makedirs(salida, exist_ok=True)
    X = savgol_filter(snv(X_raw), **SG, delta=np.mean(np.diff(wl)), axis=1, mode="interp")

    gamma0 = 1.0 / (X.shape[1] * X.var())   # equivale a gamma="scale" (sin StandardScaler)
    grid_svm = [
        {"kernel": ["rbf"], "C": C_GRID, "gamma": [gamma0 * f for f in GAMMA_FACTORES]},
        {"kernel": ["linear"], "C": C_GRID},     # control lineal
    ]
    modelos = {
        "PLS-DA": (PLSDA(), {"n_components": VL_GRID}),
        "SVM": (SVC(), grid_svm),
    }
    fijos = {"PLS-DA": {"n_components": 3},
             "SVM": {"kernel": "rbf", "C": 1.0, "gamma": gamma0}}   # si no hay CV interna posible

    print(f"\nMatriz: {X.shape} | clases: {len(np.unique(y))} | grupos: {len(np.unique(g))}")
    pred, elegidos, n_ext = validar(X, y, g, modelos, fijos)

    clases = np.unique(y)
    filas_res = []
    for nombre in modelos:
        r, por_clase, cm = metricas(y, pred[nombre], g, clases, nivel)
        filas_res.append({"Modelo": nombre, **r})
        por_clase.to_csv(os.path.join(salida, f"sens_esp_{nombre}.csv"), index=False)

        ConfusionMatrixDisplay(cm, display_labels=clases).plot(cmap="Blues", values_format="d")
        plt.title(f"Matriz de confusión (validación por lote) - {nombre}")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.savefig(os.path.join(salida, f"confusion_validacion_{nombre}.png"), dpi=200)
        plt.show()

    resumen = pd.DataFrame(filas_res)
    print("\n" + resumen.round(3).to_string(index=False))

    resumen.to_csv(os.path.join(salida, "resumen_metricas.csv"), index=False)
    elegidos.to_csv(os.path.join(salida, "hiperparametros_por_fold.csv"), index=False)
    pd.DataFrame({"archivo": trazas["archivo"].values,
                  "fila_en_archivo": trazas["fila_en_archivo"].values,
                  "clase_real": y, "grupo": g,
                  **{f"pred_{n}": pred[n] for n in modelos}}
                 ).to_csv(os.path.join(salida, "predicciones_validacion.csv"), index=False)

    config = {"SG": SG, "VL_GRID": VL_GRID, "C_GRID": C_GRID,
              "GAMMA_FACTORES": GAMMA_FACTORES, "gamma_scale_equivalente": gamma0,
              "folds_externos": n_ext, "semilla": SEMILLA,
              "escalado": False, "corte_espectral": False,
              "rango_nm": [float(wl.min()), float(wl.max())], **extra}
    with open(os.path.join(salida, "configuracion.json"), "w", encoding="utf-8") as f:
        json.dump(config, f, indent=2, ensure_ascii=False)
    print(f"[OK] Resultados en: {salida}")


if __name__ == "__main__":
    X_raw, y, g, wl, info = cargar()
    SALIDA = os.path.join(info["carpeta"], "resultados_validacion_por_lote")
    os.makedirs(SALIDA, exist_ok=True)
    info["tabla_clases"].to_csv(os.path.join(SALIDA, "clases_y_lotes.csv"))
    info["mapa"].to_csv(os.path.join(SALIDA, "mapa_archivos.csv"), index=False)

    apis = sorted(set(info["api"]))
    modo = "1"
    if len(apis) > 1:
        print(f"\nAPIs detectados: {apis}")
        modo = pedir_texto("2 = un modelo independiente por API (recomendado: el API es el "
                           "nivel principal), 1 = un solo modelo global", "2")

    extra = {"modo": "por_API" if modo == "2" else "global", "apis": apis,
             "archivos": info["archivos"],
             "criterio_clase": "campos 2-3 del código (PRODUCTO-FABRICANTE) dentro de cada API (campo 1)",
             "criterio_lote": "3 dígitos del campo 5 del código",
             "columna_unidad": info["col_unidad"]}

    if modo == "2":
        for a in apis:
            m = info["api"] == a
            print("\n" + "=" * 70 + f"\nAPI {a}\n" + "=" * 70)
            if len(np.unique(y[m])) < 2:
                print(f"[AVISO] API {a}: menos de 2 clases; se omite.")
                continue
            ejecutar(X_raw[m], y[m], g[m], info["nivel"][m], wl,
                     os.path.join(SALIDA, f"API_{a}"), extra, info["trazas"][m])
    else:
        ejecutar(X_raw, y, g, info["nivel"], wl, SALIDA, extra, info["trazas"])
```

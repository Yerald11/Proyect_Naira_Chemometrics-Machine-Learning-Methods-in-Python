```python
# ===========================================================================
# PLS-DA INTERACTIVO MEJORADO - SOLO ENTRENAMIENTO / CALIBRACIÓN
# Flujo:
# CSV -> SNV -> Savitzky-Golay -> PLS-DA preliminar
# -> diagnóstico -> eliminación manual -> PLS-DA final
# ===========================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from scipy.stats import f
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import LabelBinarizer
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

plt.style.use("default")


# ===========================================================================
# FUNCIONES DE ENTRADA
# ===========================================================================

def pedir_texto(mensaje, default=None):
    if default is None:
        return input(mensaje).strip()
    txt = input(f"{mensaje} [{default}]: ").strip()
    return default if txt == "" else txt


def pedir_entero(mensaje, default=None, minimo=None):
    while True:
        txt = pedir_texto(mensaje, default)
        try:
            valor = int(txt)
            if minimo is not None and valor < minimo:
                print(f"Debe ser >= {minimo}")
                continue
            return valor
        except ValueError:
            print("Ingresa un número entero válido.")


def pedir_si_no(mensaje, default="s"):
    while True:
        txt = pedir_texto(f"{mensaje} (s/n)", default).lower()
        if txt in ["s", "si", "sí", "y", "yes"]:
            return True
        if txt in ["n", "no"]:
            return False
        print("Responde con s o n.")


def pedir_confianza():
    while True:
        txt = pedir_texto("Nivel de confianza para diagnóstico T²: escribe 95 o 99", "95")
        if txt in ["95", "0.95"]:
            return 0.95
        if txt in ["99", "0.99"]:
            return 0.99
        print("Solo se permite 95 o 99.")


# ===========================================================================
# PREPROCESAMIENTO
# ===========================================================================

def listar_csv(carpeta):
    archivos = [f for f in os.listdir(carpeta) if f.lower().endswith(".csv")]
    archivos.sort()
    return archivos


def aplicar_snv(X, ddof=0):
    media = X.mean(axis=1, keepdims=True)
    std = X.std(axis=1, ddof=ddof, keepdims=True)

    if np.any(std == 0):
        raise ValueError("Hay espectros con desviación estándar cero. No se puede aplicar SNV.")

    return (X - media) / std


def aplicar_savgol(X, longitudes_onda, window_length, polyorder, deriv):
    if window_length % 2 == 0:
        raise ValueError("La ventana Savitzky-Golay debe ser impar.")

    if window_length <= polyorder:
        raise ValueError("La ventana debe ser mayor que el orden del polinomio.")

    diffs = np.diff(longitudes_onda)

    if not np.allclose(diffs, diffs[0]):
        print("[ADVERTENCIA] Longitudes de onda no perfectamente equiespaciadas. Se usará delta promedio.")

    delta = np.mean(diffs)

    X_sg = savgol_filter(
        X,
        window_length=window_length,
        polyorder=polyorder,
        deriv=deriv,
        delta=delta,
        axis=1,
        mode="interp"
    )

    return X_sg, delta


# ===========================================================================
# PLS-DA Y DIAGNÓSTICOS
# ===========================================================================

def calcular_vip(pls):
    T = pls.x_scores_
    W = pls.x_weights_
    Q = pls.y_loadings_

    p, h = W.shape
    ss_y = np.zeros(h)

    for a in range(h):
        t_a = T[:, a]
        q_a = Q[:, a]
        ss_y[a] = np.sum(t_a ** 2) * np.sum(q_a ** 2)

    total_ss_y = np.sum(ss_y)

    if total_ss_y == 0:
        return np.zeros(p)

    vip = np.sqrt(p * ((W ** 2) @ ss_y) / total_ss_y)
    return vip


def entrenar_pls_da(X, y, n_components):
    lb = LabelBinarizer()
    Y = lb.fit_transform(y)

    if Y.ndim == 1:
        Y = Y.reshape(-1, 1)

    n_components = min(n_components, X.shape[0] - 1, X.shape[1])

    pls = PLSRegression(n_components=n_components, scale=False)
    pls.fit(X, Y)

    Y_pred_cont = pls.predict(X)
    clases = lb.classes_

    if Y_pred_cont.shape[1] == 1:
        y_pred = np.where(Y_pred_cont.ravel() >= 0.5, clases[1], clases[0])
    else:
        y_pred = clases[np.argmax(Y_pred_cont, axis=1)]

    vip = calcular_vip(pls)

    return {
        "modelo": pls,
        "label_binarizer": lb,
        "Y": Y,
        "Y_pred_cont": Y_pred_cont,
        "y_pred": y_pred,
        "clases": clases,
        "scores": pls.x_scores_,
        "loadings": pls.x_loadings_,
        "weights": pls.x_weights_,
        "coef": pls.coef_,
        "vip": vip,
        "x_mean": X.mean(axis=0)
    }


def diagnostico_plsda(X, plsda_res, alpha=0.95):
    pls = plsda_res["modelo"]
    T = plsda_res["scores"]
    P = pls.x_loadings_
    Y = plsda_res["Y"]
    Y_pred = plsda_res["Y_pred_cont"]

    n, a = T.shape

    var_scores = np.var(T, axis=0, ddof=1)
    var_scores = np.where(var_scores == 0, np.nan, var_scores)

    T2 = np.nansum((T ** 2) / var_scores, axis=1)

    if n > a:
        T2_lim = a * (n - 1) / (n - a) * f.ppf(alpha, a, n - a)
    else:
        T2_lim = np.nan

    Xc = X - X.mean(axis=0)
    X_hat = T @ P.T
    QX = np.sum((Xc - X_hat) ** 2, axis=1)

    EY = np.sqrt(np.sum((Y - Y_pred) ** 2, axis=1))

    return T2, QX, EY, T2_lim


def graficar_diagnostico_plsda(plsda_res, y_labels, T2, QX, EY, T2_lim, alpha, titulo="preliminar"):
    scores = plsda_res["scores"]
    confianza = int(alpha * 100)

    fig, ax = plt.subplots(1, 4, figsize=(23, 5))

    if scores.shape[1] >= 2:
        x_score = scores[:, 0]
        y_score = scores[:, 1]
        xlabel = "PLS1"
        ylabel = "PLS2"
    else:
        x_score = scores[:, 0]
        y_score = np.zeros(scores.shape[0])
        xlabel = "PLS1"
        ylabel = "0"

    for clase in np.unique(y_labels):
        idx = y_labels == clase
        ax[0].scatter(
            x_score[idx],
            y_score[idx],
            label=clase,
            s=65,
            edgecolors="black",
            linewidths=0.6,
            alpha=0.85
        )

    for i in range(scores.shape[0]):
        ax[0].text(x_score[i], y_score[i], str(i), fontsize=8, alpha=0.7)

    ax[0].axhline(0, color="black", linestyle=":", alpha=0.5)
    ax[0].axvline(0, color="black", linestyle=":", alpha=0.5)
    ax[0].set_xlabel(xlabel)
    ax[0].set_ylabel(ylabel)
    ax[0].set_title(f"Scores PLS-DA {titulo}")
    ax[0].legend()
    ax[0].grid(alpha=0.2)

    colores_t2 = np.where(T2 > T2_lim, "crimson", "teal")
    ax[1].bar(range(len(T2)), T2, color=colores_t2, alpha=0.8)
    ax[1].axhline(T2_lim, color="red", linestyle="--", label=f"Límite {confianza}%")
    ax[1].set_title(f"Hotelling T² {titulo}")
    ax[1].set_xlabel("Índice")
    ax[1].set_ylabel("T²")
    ax[1].legend()
    ax[1].grid(axis="y", alpha=0.25)

    ax[2].bar(range(len(QX)), QX, color="slateblue", alpha=0.8)
    ax[2].set_title(f"Q residual X {titulo}")
    ax[2].set_xlabel("Índice")
    ax[2].set_ylabel("QX")
    ax[2].grid(axis="y", alpha=0.25)

    ax[3].bar(range(len(EY)), EY, color="darkorange", alpha=0.8)
    ax[3].set_title(f"Error residual Y {titulo}")
    ax[3].set_xlabel("Índice")
    ax[3].set_ylabel("EY")
    ax[3].grid(axis="y", alpha=0.25)

    plt.tight_layout()
    plt.show()


def graficar_resultados_plsda(X_raw, X_proc, longitudes_onda, plsda_res, y_labels, titulo="final"):
    scores = plsda_res["scores"]
    vip = plsda_res["vip"]
    y_pred = plsda_res["y_pred"]
    clases = plsda_res["clases"]
    Y_pred_cont = plsda_res["Y_pred_cont"]

    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    ax[0].plot(longitudes_onda, X_raw.T, alpha=0.45)
    ax[0].set_title("Espectros originales")
    ax[0].set_xlabel("Longitud de onda")
    ax[0].set_ylabel("Absorbancia")
    ax[0].grid(alpha=0.2)

    ax[1].plot(longitudes_onda, X_proc.T, alpha=0.45)
    ax[1].set_title("Espectros preprocesados")
    ax[1].set_xlabel("Longitud de onda")
    ax[1].set_ylabel("Intensidad preprocesada")
    ax[1].grid(alpha=0.2)

    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 6))

    if scores.shape[1] >= 2:
        xs = scores[:, 0]
        ys = scores[:, 1]
        ylabel = "PLS2"
    else:
        xs = scores[:, 0]
        ys = np.zeros(scores.shape[0])
        ylabel = "0"

    for clase in np.unique(y_labels):
        idx = y_labels == clase
        plt.scatter(
            xs[idx],
            ys[idx],
            label=clase,
            s=65,
            edgecolors="black",
            linewidths=0.6,
            alpha=0.85
        )

    for i in range(scores.shape[0]):
        plt.text(xs[i], ys[i], str(i), fontsize=8, alpha=0.65)

    plt.axhline(0, color="black", linestyle=":", alpha=0.5)
    plt.axvline(0, color="black", linestyle=":", alpha=0.5)
    plt.xlabel("PLS1")
    plt.ylabel(ylabel)
    plt.title(f"Scores PLS-DA - {titulo}")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    cm = confusion_matrix(y_labels, y_pred, labels=clases)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=clases)
    disp.plot(cmap="Blues", values_format="d")
    plt.title(f"Matriz de confusión aparente - {titulo}")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 4))
    plt.plot(longitudes_onda, vip, color="teal", linewidth=1.5)
    plt.axhline(1, color="red", linestyle="--", label="VIP = 1")
    plt.xlabel("Longitud de onda")
    plt.ylabel("VIP")
    plt.title(f"VIP scores - {titulo}")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))

    if Y_pred_cont.shape[1] > 1:
        for i, clase in enumerate(clases):
            plt.plot(Y_pred_cont[:, i], marker="o", linestyle="", label=f"Respuesta {clase}")
        plt.axhline(0.5, color="red", linestyle="--", label="Umbral 0.5")
    else:
        plt.plot(Y_pred_cont.ravel(), marker="o", linestyle="", label="Respuesta binaria")
        plt.axhline(0.5, color="red", linestyle="--", label="Umbral 0.5")

    plt.xlabel("Índice de muestra")
    plt.ylabel("Respuesta PLS-DA")
    plt.title(f"Respuestas continuas PLS-DA - {titulo}")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()


# ===========================================================================
# 1. CONFIGURACIÓN INTERACTIVA
# ===========================================================================

print("=" * 75)
print("PLS-DA INTERACTIVO MEJORADO - SOLO ENTRENAMIENTO / CALIBRACIÓN")
print("Flujo: SNV + Savitzky-Golay + PLS-DA preliminar + descarte manual opcional")
print("=" * 75)

carpeta = pedir_texto("\nIngresa la carpeta donde están los CSV", ".")
carpeta = os.path.abspath(carpeta)

if not os.path.isdir(carpeta):
    raise FileNotFoundError(f"No existe la carpeta: {carpeta}")

csv_disponibles = listar_csv(carpeta)

if len(csv_disponibles) == 0:
    raise FileNotFoundError("No se encontraron archivos CSV.")

print("\nCSV encontrados:")
for i, archivo in enumerate(csv_disponibles, start=1):
    print(f"{i}. {archivo}")

seleccion_txt = pedir_texto("\nEscribe los números de los CSV separados por coma. Ejemplo: 1,2,3")
indices_csv = [int(x.strip()) - 1 for x in seleccion_txt.split(",")]
archivos_seleccionados = [csv_disponibles[i] for i in indices_csv]

etiquetas_clase = {}

print("\nAsigna una etiqueta/clase para cada archivo.")
for archivo in archivos_seleccionados:
    etiqueta_default = os.path.splitext(archivo)[0]
    etiqueta = pedir_texto(f"Etiqueta para {archivo}", etiqueta_default)
    etiquetas_clase[archivo] = etiqueta

sg_window = pedir_entero("\nVentana Savitzky-Golay", default=17, minimo=3)
sg_polyorder = pedir_entero("Orden del polinomio Savitzky-Golay", default=2, minimo=1)
sg_deriv = pedir_entero("Orden de derivada Savitzky-Golay", default=1, minimo=0)

n_components = pedir_entero("\nNúmero de componentes PLS", default=3, minimo=1)
alpha = pedir_confianza()

exportar = pedir_si_no("\n¿Deseas exportar resultados CSV?", default="s")


# ===========================================================================
# 2. CARGA DE DATOS
# ===========================================================================

dfs = []
y_labels = []

for archivo in archivos_seleccionados:
    ruta = os.path.join(carpeta, archivo)
    df = pd.read_csv(ruta)
    clase = etiquetas_clase[archivo]

    dfs.append(df)
    y_labels.extend([clase] * len(df))

    print(f"[OK] {archivo}: {df.shape[0]} muestras, clase = {clase}")

df_global = pd.concat(dfs, ignore_index=True)
y_labels = np.array(y_labels)

columnas_convertidas = pd.to_numeric(df_global.columns, errors="coerce")
columnas_espectrales = df_global.columns[columnas_convertidas.notna()]
columnas_metadata = df_global.columns[columnas_convertidas.isna()]

longitudes_onda = columnas_convertidas[columnas_convertidas.notna()].astype(float).values
X_raw = df_global[columnas_espectrales].astype(float).values

orden = np.argsort(longitudes_onda)
longitudes_onda = longitudes_onda[orden]
X_raw = X_raw[:, orden]

print("\nColumnas de metadatos omitidas:")
print(list(columnas_metadata))

print(f"\nMatriz espectral cruda: {X_raw.shape} = muestras x variables")

if np.isnan(X_raw).any():
    raise ValueError("La matriz contiene NaN.")

if np.isinf(X_raw).any():
    raise ValueError("La matriz contiene infinitos.")


# ===========================================================================
# 3. PREPROCESAMIENTO
# ===========================================================================

X_snv = aplicar_snv(X_raw, ddof=0)

X_proc, delta_sg = aplicar_savgol(
    X_snv,
    longitudes_onda,
    window_length=sg_window,
    polyorder=sg_polyorder,
    deriv=sg_deriv
)

print("\nPreprocesamiento aplicado:")
print("- SNV muestra por muestra, ddof=0")
print(f"- Savitzky-Golay: ventana={sg_window}, polinomio={sg_polyorder}, derivada={sg_deriv}, delta={delta_sg:.6f}")


# ===========================================================================
# 4. PLS-DA PRELIMINAR
# ===========================================================================

plsda_pre = entrenar_pls_da(X_proc, y_labels, n_components)
T2_pre, QX_pre, EY_pre, T2_lim_pre = diagnostico_plsda(X_proc, plsda_pre, alpha=alpha)

print("\n" + "=" * 75)
print("PLS-DA PRELIMINAR - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels, plsda_pre["y_pred"]))

graficar_resultados_plsda(
    X_raw,
    X_proc,
    longitudes_onda,
    plsda_pre,
    y_labels,
    titulo="preliminar"
)

graficar_diagnostico_plsda(
    plsda_pre,
    y_labels,
    T2_pre,
    QX_pre,
    EY_pre,
    T2_lim_pre,
    alpha,
    titulo="preliminar"
)

df_diag_pre = pd.DataFrame({
    "Indice": np.arange(len(y_labels)),
    "Clase_real": y_labels,
    "Clase_predicha": plsda_pre["y_pred"],
    "Correcta": y_labels == plsda_pre["y_pred"],
    "T2": T2_pre,
    "QX_residual": QX_pre,
    "Error_Y": EY_pre,
    "T2_fuera_limite": T2_pre > T2_lim_pre
})

print("\nMuestras con mayor Hotelling T²:")
display(df_diag_pre.sort_values("T2", ascending=False).head(15))

print("\nMuestras con mayor Q residual X:")
display(df_diag_pre.sort_values("QX_residual", ascending=False).head(15))

print("\nMuestras con mayor error residual Y:")
display(df_diag_pre.sort_values("Error_Y", ascending=False).head(15))

print("\nMuestras mal clasificadas en calibración aparente:")
display(df_diag_pre[df_diag_pre["Correcta"] == False])


# ===========================================================================
# 5. ELIMINACIÓN MANUAL DESPUÉS DEL DIAGNÓSTICO
# ===========================================================================

eliminar = pedir_si_no(
    "\nDespués de ver los diagnósticos PLS-DA, ¿deseas eliminar muestras manualmente?",
    default="n"
)

indices_eliminar = []

if eliminar:
    txt = pedir_texto("Escribe los índices globales a eliminar separados por coma. Ejemplo: 54 o 54,67")
    if txt.strip() != "":
        indices_eliminar = [int(x.strip()) for x in txt.split(",")]

if len(indices_eliminar) > 0:
    indices_eliminar = np.array(indices_eliminar, dtype=int)

    if np.any(indices_eliminar < 0) or np.any(indices_eliminar >= X_raw.shape[0]):
        raise ValueError("Hay índices a eliminar fuera del rango.")

    mascara = np.ones(X_raw.shape[0], dtype=bool)
    mascara[indices_eliminar] = False

    print("\nMuestras eliminadas manualmente:")
    for idx in indices_eliminar:
        print(f"- Índice global {idx}, clase {y_labels[idx]}")

    X_raw_final = X_raw[mascara]
    X_proc_final = X_proc[mascara]
    y_labels_final = y_labels[mascara]
    df_global_final = df_global.loc[mascara].reset_index(drop=True)

else:
    print("\nNo se eliminaron muestras. El PLS-DA final usará todas las muestras.")
    X_raw_final = X_raw.copy()
    X_proc_final = X_proc.copy()
    y_labels_final = y_labels.copy()
    df_global_final = df_global.copy()

print(f"\nMatriz final para PLS-DA: {X_proc_final.shape}")


# ===========================================================================
# 6. PLS-DA FINAL
# ===========================================================================

plsda_final = entrenar_pls_da(X_proc_final, y_labels_final, n_components)
T2_final, QX_final, EY_final, T2_lim_final = diagnostico_plsda(X_proc_final, plsda_final, alpha=alpha)

print("\n" + "=" * 75)
print("PLS-DA FINAL - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels_final, plsda_final["y_pred"]))

graficar_resultados_plsda(
    X_raw_final,
    X_proc_final,
    longitudes_onda,
    plsda_final,
    y_labels_final,
    titulo="final"
)

graficar_diagnostico_plsda(
    plsda_final,
    y_labels_final,
    T2_final,
    QX_final,
    EY_final,
    T2_lim_final,
    alpha,
    titulo="final"
)


# ===========================================================================
# 7. TABLAS FINALES
# ===========================================================================

df_pred_final = pd.DataFrame({
    "Indice": np.arange(len(y_labels_final)),
    "Clase_real": y_labels_final,
    "Clase_predicha_calibracion": plsda_final["y_pred"],
    "Correcta": y_labels_final == plsda_final["y_pred"],
    "T2": T2_final,
    "QX_residual": QX_final,
    "Error_Y": EY_final,
    "T2_fuera_limite": T2_final > T2_lim_final
})

Y_pred_cont = plsda_final["Y_pred_cont"]

if Y_pred_cont.shape[1] > 1:
    for i, clase in enumerate(plsda_final["clases"]):
        df_pred_final[f"Respuesta_{clase}"] = Y_pred_cont[:, i]
else:
    df_pred_final["Respuesta_binaria"] = Y_pred_cont.ravel()

display(df_pred_final)

df_vip = pd.DataFrame({
    "Longitud_onda": longitudes_onda,
    "VIP": plsda_final["vip"]
}).sort_values("VIP", ascending=False)

print("\nVariables con mayor VIP:")
display(df_vip.head(20))


# ===========================================================================
# 8. EXPORTACIÓN
# ===========================================================================

if exportar:
    carpeta_salida = os.path.join(carpeta, "resultados_pls_da_entrenamiento_mejorado")
    os.makedirs(carpeta_salida, exist_ok=True)

    scores = plsda_final["scores"]

    df_scores = pd.DataFrame(
        scores,
        columns=[f"PLS{i+1}" for i in range(scores.shape[1])]
    )
    df_scores.insert(0, "Clase", y_labels_final)
    df_scores.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_loadings = pd.DataFrame(
        plsda_final["loadings"],
        columns=[f"PLS{i+1}" for i in range(plsda_final["loadings"].shape[1])]
    )
    df_loadings.insert(0, "Longitud_onda", longitudes_onda)

    df_weights = pd.DataFrame(
        plsda_final["weights"],
        columns=[f"PLS{i+1}" for i in range(plsda_final["weights"].shape[1])]
    )
    df_weights.insert(0, "Longitud_onda", longitudes_onda)

    df_preprocesado = pd.DataFrame(X_proc_final, columns=[str(w) for w in longitudes_onda])
    df_preprocesado.insert(0, "Clase", y_labels_final)
    df_preprocesado.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_eliminadas = pd.DataFrame({
        "Indice_original_eliminado": indices_eliminar
    })

    df_pred_final.to_csv(os.path.join(carpeta_salida, "plsda_predicciones_calibracion_final.csv"), index=False)
    df_scores.to_csv(os.path.join(carpeta_salida, "plsda_scores_final.csv"), index=False)
    df_loadings.to_csv(os.path.join(carpeta_salida, "plsda_loadings_final.csv"), index=False)
    df_weights.to_csv(os.path.join(carpeta_salida, "plsda_weights_final.csv"), index=False)
    df_vip.to_csv(os.path.join(carpeta_salida, "plsda_vip_final.csv"), index=False)
    df_preprocesado.to_csv(os.path.join(carpeta_salida, "matriz_preprocesada_snv_sg_final.csv"), index=False)
    df_diag_pre.to_csv(os.path.join(carpeta_salida, "diagnostico_plsda_preliminar.csv"), index=False)
    df_eliminadas.to_csv(os.path.join(carpeta_salida, "muestras_eliminadas.csv"), index=False)

    print("\n[OK] Resultados exportados en:")
    print(carpeta_salida)
```

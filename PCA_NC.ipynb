```python
# ===========================================================================
# PCA QUIMIOMÉTRICO INTERACTIVO - PCA NO CENTRADO
# Flujo:
# CSV -> SNV -> Savitzky-Golay -> PCA no centrado preliminar
# -> diagnóstico -> eliminación manual -> PCA no centrado final
# ===========================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from scipy.stats import f

plt.style.use("default")


# ===========================================================================
# FUNCIONES DE ENTRADA
# ===========================================================================

def pedir_texto(mensaje, default=None):
    if default is None:
        return input(mensaje).strip()
    txt = input(f"{mensaje} [{default}]: ").strip()
    return default if txt == "" else txt


def pedir_entero(mensaje, default=None, minimo=None):
    while True:
        txt = pedir_texto(mensaje, default)
        try:
            valor = int(txt)
            if minimo is not None and valor < minimo:
                print(f"Debe ser >= {minimo}")
                continue
            return valor
        except ValueError:
            print("Ingresa un número entero válido.")


def pedir_si_no(mensaje, default="n"):
    while True:
        txt = pedir_texto(f"{mensaje} (s/n)", default).lower()
        if txt in ["s", "si", "sí", "y", "yes"]:
            return True
        if txt in ["n", "no"]:
            return False
        print("Responde con s o n.")


def pedir_confianza():
    while True:
        txt = pedir_texto("Nivel de confianza para Hotelling T²: escribe 95 o 99", "95")
        if txt in ["95", "0.95"]:
            return 0.95
        if txt in ["99", "0.99"]:
            return 0.99
        print("Solo se permite 95 o 99.")


# ===========================================================================
# FUNCIONES QUIMIOMÉTRICAS
# ===========================================================================

def listar_csv(carpeta):
    archivos = [f for f in os.listdir(carpeta) if f.lower().endswith(".csv")]
    archivos.sort()
    return archivos


def aplicar_snv(X, ddof=0):
    media = X.mean(axis=1, keepdims=True)
    std = X.std(axis=1, ddof=ddof, keepdims=True)

    if np.any(std == 0):
        raise ValueError("Hay espectros con desviación estándar cero. No se puede aplicar SNV.")

    return (X - media) / std


def aplicar_savgol(X, longitudes_onda, window_length, polyorder, deriv):
    if window_length % 2 == 0:
        raise ValueError("La ventana Savitzky-Golay debe ser impar.")

    if window_length <= polyorder:
        raise ValueError("La ventana debe ser mayor que el orden del polinomio.")

    diffs = np.diff(longitudes_onda)

    if not np.allclose(diffs, diffs[0]):
        print("[ADVERTENCIA] Las longitudes de onda no son perfectamente equiespaciadas.")
        print("[ADVERTENCIA] Se usará delta promedio.")

    delta = np.mean(diffs)

    X_sg = savgol_filter(
        X,
        window_length=window_length,
        polyorder=polyorder,
        deriv=deriv,
        delta=delta,
        axis=1,
        mode="interp"
    )

    return X_sg, delta


def calcular_pca_no_centrado(X, n_components):
    n_components = min(n_components, X.shape[0], X.shape[1])

    U, S, Vt = np.linalg.svd(X, full_matrices=False)

    scores = U[:, :n_components] * S[:n_components]
    loadings = Vt[:n_components, :]

    var_ratio = (S[:n_components] ** 2) / np.sum(S ** 2) * 100
    var_acum = np.cumsum(var_ratio)

    # Referencia tipo eigenvalue. En PCA no centrado no corresponde exactamente
    # a eigenvalues de matriz de covarianza centrada.
    eigenvalues = (S[:n_components] ** 2) / (X.shape[0] - 1)

    return {
        "scores": scores,
        "loadings": loadings,
        "var_ratio": var_ratio,
        "var_acum": var_acum,
        "eigenvalues": eigenvalues,
        "singular_values": S[:n_components],
        "mean": np.zeros(X.shape[1])
    }


def diagnostico_t2_q_no_centrado(X, pca_res, n_comp_diag=3, alpha=0.95):
    scores = pca_res["scores"][:, :n_comp_diag]
    loadings = pca_res["loadings"][:n_comp_diag, :]
    eigenvalues = pca_res["eigenvalues"][:n_comp_diag]

    X_hat = scores @ loadings
    residuos = X - X_hat

    Q = np.sum(residuos ** 2, axis=1)
    T2 = np.sum((scores ** 2) / eigenvalues, axis=1)

    n = X.shape[0]
    a = n_comp_diag

    if n > a:
        T2_lim = a * (n - 1) / (n - a) * f.ppf(alpha, a, n - a)
    else:
        T2_lim = np.nan

    return T2, Q, T2_lim


# ===========================================================================
# FUNCIONES DE GRÁFICAS
# ===========================================================================

def graficar_diagnostico(scores, var_ratio, y_labels, T2, Q, T2_lim, alpha, titulo_extra=""):
    confianza = int(alpha * 100)

    fig, ax = plt.subplots(1, 3, figsize=(19, 5))

    for clase in np.unique(y_labels):
        idx = y_labels == clase
        ax[0].scatter(
            scores[idx, 0],
            scores[idx, 1],
            label=clase,
            s=65,
            edgecolors="black",
            linewidths=0.6,
            alpha=0.85
        )

    for i in range(scores.shape[0]):
        ax[0].text(scores[i, 0], scores[i, 1], str(i), fontsize=8, alpha=0.75)

    ax[0].axhline(0, color="black", linestyle=":", alpha=0.5)
    ax[0].axvline(0, color="black", linestyle=":", alpha=0.5)
    ax[0].set_xlabel(f"PC1 ({var_ratio[0]:.2f}%)")
    ax[0].set_ylabel(f"PC2 ({var_ratio[1]:.2f}%)")
    ax[0].set_title(f"Scores PC1 vs PC2 {titulo_extra}")
    ax[0].legend()
    ax[0].grid(alpha=0.2)

    colores_t2 = np.where(T2 > T2_lim, "crimson", "teal")

    ax[1].bar(range(len(T2)), T2, color=colores_t2, alpha=0.8)
    ax[1].axhline(
        T2_lim,
        color="red",
        linestyle="--",
        label=f"Límite {confianza}% = {T2_lim:.2f}"
    )
    ax[1].set_title(f"Hotelling T² {titulo_extra}")
    ax[1].set_xlabel("Índice de muestra")
    ax[1].set_ylabel("T²")
    ax[1].legend()
    ax[1].grid(axis="y", alpha=0.25)

    ax[2].bar(range(len(Q)), Q, color="slateblue", alpha=0.8)
    ax[2].set_title(f"Q residual {titulo_extra}")
    ax[2].set_xlabel("Índice de muestra")
    ax[2].set_ylabel("Q residual")
    ax[2].grid(axis="y", alpha=0.25)

    plt.tight_layout()
    plt.show()


def graficar_resultados_finales(X_raw, X_proc, longitudes_onda, pca_res, y_labels, titulo="PCA no centrado final"):
    scores = pca_res["scores"]
    loadings = pca_res["loadings"]
    var_ratio = pca_res["var_ratio"]
    var_acum = pca_res["var_acum"]

    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    ax[0].plot(longitudes_onda, X_raw.T, alpha=0.45)
    ax[0].set_title("Espectros originales")
    ax[0].set_xlabel("Longitud de onda")
    ax[0].set_ylabel("Absorbancia")
    ax[0].grid(alpha=0.2)

    ax[1].plot(longitudes_onda, X_proc.T, alpha=0.45)
    ax[1].set_title("Espectros preprocesados")
    ax[1].set_xlabel("Longitud de onda")
    ax[1].set_ylabel("Intensidad preprocesada")
    ax[1].grid(alpha=0.2)

    plt.tight_layout()
    plt.show()

    n_plot = min(10, len(var_ratio))

    plt.figure(figsize=(8, 4))
    bars = plt.bar(range(1, n_plot + 1), var_ratio[:n_plot], color="teal", alpha=0.75)
    plt.step(range(1, n_plot + 1), var_acum[:n_plot], where="mid", color="red", linestyle="--")

    for bar in bars:
        yval = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            yval + 0.5,
            f"{yval:.2f}%",
            ha="center",
            fontsize=8
        )

    plt.xlabel("Componentes principales")
    plt.ylabel("Varianza explicada / energía (%)")
    plt.title(f"Scree Plot - {titulo}")
    plt.xticks(range(1, n_plot + 1), [f"PC{i}" for i in range(1, n_plot + 1)])
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 6))

    for clase in np.unique(y_labels):
        idx = y_labels == clase
        plt.scatter(
            scores[idx, 0],
            scores[idx, 1],
            label=clase,
            s=65,
            edgecolors="black",
            linewidths=0.6,
            alpha=0.85
        )

    for i in range(scores.shape[0]):
        plt.text(scores[i, 0], scores[i, 1], str(i), fontsize=8, alpha=0.7)

    plt.axhline(0, color="black", linestyle=":", alpha=0.5)
    plt.axvline(0, color="black", linestyle=":", alpha=0.5)
    plt.xlabel(f"PC1 ({var_ratio[0]:.2f}%)")
    plt.ylabel(f"PC2 ({var_ratio[1]:.2f}%)")
    plt.title(f"Scores PC1 vs PC2 - {titulo}")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))
    plt.plot(longitudes_onda, loadings[0], label=f"PC1 ({var_ratio[0]:.2f}%)")
    plt.plot(longitudes_onda, loadings[1], label=f"PC2 ({var_ratio[1]:.2f}%)")
    plt.axhline(0, color="black", linestyle=":", alpha=0.5)
    plt.xlabel("Longitud de onda")
    plt.ylabel("Loading")
    plt.title(f"Loadings PC1 y PC2 - {titulo}")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()


# ===========================================================================
# 1. CONFIGURACIÓN INTERACTIVA
# ===========================================================================

print("=" * 75)
print("ANÁLISIS PCA QUIMIOMÉTRICO INTERACTIVO - PCA NO CENTRADO")
print("Flujo: SNV + Savitzky-Golay + PCA no centrado")
print("=" * 75)

carpeta = pedir_texto("\nIngresa la carpeta donde están los CSV", ".")
carpeta = os.path.abspath(carpeta)

if not os.path.isdir(carpeta):
    raise FileNotFoundError(f"No existe la carpeta: {carpeta}")

csv_disponibles = listar_csv(carpeta)

if len(csv_disponibles) == 0:
    raise FileNotFoundError("No se encontraron archivos CSV en la carpeta indicada.")

print("\nCSV encontrados:")
for i, archivo in enumerate(csv_disponibles, start=1):
    print(f"{i}. {archivo}")

seleccion_txt = pedir_texto("\nEscribe los números de los CSV que quieres usar separados por coma. Ejemplo: 1,2,3")
indices_csv = [int(x.strip()) - 1 for x in seleccion_txt.split(",")]

archivos_seleccionados = []

for idx in indices_csv:
    if idx < 0 or idx >= len(csv_disponibles):
        raise ValueError("Uno de los números seleccionados está fuera de rango.")
    archivos_seleccionados.append(csv_disponibles[idx])

print("\nArchivos seleccionados:")
for archivo in archivos_seleccionados:
    print(f"- {archivo}")

etiquetas_clase = {}

print("\nAsigna una etiqueta/clase para cada archivo.")
for archivo in archivos_seleccionados:
    etiqueta_default = os.path.splitext(archivo)[0]
    etiqueta = pedir_texto(f"Etiqueta para {archivo}", etiqueta_default)
    etiquetas_clase[archivo] = etiqueta

sg_window = pedir_entero("\nVentana Savitzky-Golay", default=17, minimo=3)
sg_polyorder = pedir_entero("Orden del polinomio Savitzky-Golay", default=2, minimo=1)
sg_deriv = pedir_entero("Orden de derivada Savitzky-Golay", default=1, minimo=0)

n_components = pedir_entero("\nNúmero de componentes principales a calcular", default=10, minimo=2)
alpha = pedir_confianza()

exportar = pedir_si_no("\n¿Deseas exportar resultados CSV?", default="s")


# ===========================================================================
# 2. CARGA DE DATOS
# ===========================================================================

dfs = []
y_labels = []

for archivo in archivos_seleccionados:
    ruta = os.path.join(carpeta, archivo)
    df = pd.read_csv(ruta)

    clase = etiquetas_clase[archivo]

    dfs.append(df)
    y_labels.extend([clase] * len(df))

    print(f"[OK] {archivo}: {df.shape[0]} muestras, clase = {clase}")

df_global = pd.concat(dfs, ignore_index=True)
y_labels = np.array(y_labels)

columnas_convertidas = pd.to_numeric(df_global.columns, errors="coerce")

columnas_espectrales = df_global.columns[columnas_convertidas.notna()]
columnas_metadata = df_global.columns[columnas_convertidas.isna()]

longitudes_onda = columnas_convertidas[columnas_convertidas.notna()].astype(float).values
X_raw = df_global[columnas_espectrales].astype(float).values

orden = np.argsort(longitudes_onda)
longitudes_onda = longitudes_onda[orden]
X_raw = X_raw[:, orden]

print("\nColumnas de metadatos omitidas:")
print(list(columnas_metadata))

print(f"\nMatriz espectral cruda: {X_raw.shape} = muestras x variables")
print(f"Rango espectral: {longitudes_onda.min()} - {longitudes_onda.max()}")

if np.isnan(X_raw).any():
    raise ValueError("La matriz contiene NaN. Revisa los CSV.")

if np.isinf(X_raw).any():
    raise ValueError("La matriz contiene valores infinitos.")


# ===========================================================================
# 3. PREPROCESAMIENTO
# ===========================================================================

X_snv = aplicar_snv(X_raw, ddof=0)

X_proc, delta_sg = aplicar_savgol(
    X_snv,
    longitudes_onda,
    window_length=sg_window,
    polyorder=sg_polyorder,
    deriv=sg_deriv
)

print("\nPreprocesamiento aplicado:")
print("- SNV muestra por muestra, ddof=0")
print(f"- Savitzky-Golay: ventana={sg_window}, polinomio={sg_polyorder}, derivada={sg_deriv}, delta={delta_sg:.6f}")
print(f"Matriz preprocesada: {X_proc.shape}")


# ===========================================================================
# 4. PCA NO CENTRADO PRELIMINAR Y DIAGNÓSTICO
# ===========================================================================

pca_pre = calcular_pca_no_centrado(X_proc, n_components)

n_comp_diag = min(3, pca_pre["scores"].shape[1])

T2_pre, Q_pre, T2_lim_pre = diagnostico_t2_q_no_centrado(
    X_proc,
    pca_pre,
    n_comp_diag=n_comp_diag,
    alpha=alpha
)

print("\n" + "=" * 75)
print("PCA NO CENTRADO PRELIMINAR")
print("=" * 75)

for i in range(len(pca_pre["var_ratio"])):
    print(
        f"PC{i+1}: {pca_pre['var_ratio'][i]:.6f}% | "
        f"Acumulada: {pca_pre['var_acum'][i]:.6f}% | "
        f"Eigenvalue ref.: {pca_pre['eigenvalues'][i]:.8f}"
    )

graficar_diagnostico(
    pca_pre["scores"],
    pca_pre["var_ratio"],
    y_labels,
    T2_pre,
    Q_pre,
    T2_lim_pre,
    alpha,
    titulo_extra="preliminar no centrado"
)

df_diag_pre = pd.DataFrame({
    "Indice": np.arange(len(y_labels)),
    "Clase": y_labels,
    "T2": T2_pre,
    "Q_residual": Q_pre,
    "T2_fuera_limite": T2_pre > T2_lim_pre
})

print("\nMuestras con mayor Hotelling T²:")
display(df_diag_pre.sort_values("T2", ascending=False).head(15))

print("\nMuestras con mayor Q residual:")
display(df_diag_pre.sort_values("Q_residual", ascending=False).head(15))


# ===========================================================================
# 5. ELIMINACIÓN MANUAL DESPUÉS DE VER GRÁFICAS
# ===========================================================================

eliminar = pedir_si_no(
    "\nDespués de ver Scores, Hotelling T² y Q residual, ¿deseas eliminar muestras manualmente?",
    default="n"
)

indices_eliminar = []

if eliminar:
    txt = pedir_texto("Escribe los índices globales a eliminar separados por coma. Ejemplo: 25 o 10,25,31")
    if txt.strip() != "":
        indices_eliminar = [int(x.strip()) for x in txt.split(",")]

if len(indices_eliminar) > 0:
    indices_eliminar = np.array(indices_eliminar, dtype=int)

    if np.any(indices_eliminar < 0) or np.any(indices_eliminar >= X_raw.shape[0]):
        raise ValueError("Hay índices a eliminar fuera del rango.")

    mascara = np.ones(X_raw.shape[0], dtype=bool)
    mascara[indices_eliminar] = False

    print("\nMuestras eliminadas manualmente:")
    for idx in indices_eliminar:
        print(f"- Índice global {idx}, clase {y_labels[idx]}")

    X_raw_final = X_raw[mascara]
    X_proc_final = X_proc[mascara]
    y_labels_final = y_labels[mascara]
    df_global_final = df_global.loc[mascara].reset_index(drop=True)

else:
    print("\nNo se eliminaron muestras. El PCA final usará todas las muestras.")
    X_raw_final = X_raw.copy()
    X_proc_final = X_proc.copy()
    y_labels_final = y_labels.copy()
    df_global_final = df_global.copy()

print(f"\nMatriz final para PCA: {X_proc_final.shape}")


# ===========================================================================
# 6. PCA NO CENTRADO FINAL
# ===========================================================================

pca_final = calcular_pca_no_centrado(X_proc_final, n_components)

scores = pca_final["scores"]
loadings = pca_final["loadings"]
var_ratio = pca_final["var_ratio"]
var_acum = pca_final["var_acum"]
eigenvalues = pca_final["eigenvalues"]

T2_final, Q_final, T2_lim_final = diagnostico_t2_q_no_centrado(
    X_proc_final,
    pca_final,
    n_comp_diag=min(3, scores.shape[1]),
    alpha=alpha
)

print("\n" + "=" * 75)
print("PCA NO CENTRADO FINAL")
print("=" * 75)

for i in range(len(var_ratio)):
    print(
        f"PC{i+1}: {var_ratio[i]:.6f}% | "
        f"Acumulada: {var_acum[i]:.6f}% | "
        f"Eigenvalue ref.: {eigenvalues[i]:.8f}"
    )

graficar_resultados_finales(
    X_raw_final,
    X_proc_final,
    longitudes_onda,
    pca_final,
    y_labels_final,
    titulo="PCA no centrado final"
)

graficar_diagnostico(
    scores,
    var_ratio,
    y_labels_final,
    T2_final,
    Q_final,
    T2_lim_final,
    alpha,
    titulo_extra="final no centrado"
)

df_diagnostico_final = pd.DataFrame({
    "Indice_final": np.arange(len(y_labels_final)),
    "Clase": y_labels_final,
    "T2": T2_final,
    "Q_residual": Q_final,
    "T2_fuera_limite": T2_final > T2_lim_final
})

display(df_diagnostico_final.sort_values(["T2_fuera_limite", "T2"], ascending=[False, False]).head(15))


# ===========================================================================
# 7. EXPORTACIÓN
# ===========================================================================

if exportar:
    carpeta_salida = os.path.join(carpeta, "resultados_pca_no_centrado")
    os.makedirs(carpeta_salida, exist_ok=True)

    df_scores = pd.DataFrame(scores, columns=[f"PC{i+1}" for i in range(scores.shape[1])])
    df_scores.insert(0, "Clase", y_labels_final)
    df_scores.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_loadings = pd.DataFrame(loadings.T, columns=[f"PC{i+1}" for i in range(loadings.shape[0])])
    df_loadings.insert(0, "Longitud_onda", longitudes_onda)

    df_varianza = pd.DataFrame({
        "PC": [f"PC{i+1}" for i in range(len(var_ratio))],
        "Varianza_explicada_energia_%": var_ratio,
        "Varianza_acumulada_energia_%": var_acum,
        "Eigenvalue_referencia": eigenvalues,
        "Singular_value": pca_final["singular_values"]
    })

    df_preprocesado = pd.DataFrame(X_proc_final, columns=[str(w) for w in longitudes_onda])
    df_preprocesado.insert(0, "Clase", y_labels_final)
    df_preprocesado.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_eliminadas = pd.DataFrame({
        "Indice_original_eliminado": indices_eliminar
    })

    df_scores.to_csv(os.path.join(carpeta_salida, "scores_pca_no_centrado_final.csv"), index=False)
    df_loadings.to_csv(os.path.join(carpeta_salida, "loadings_pca_no_centrado_final.csv"), index=False)
    df_varianza.to_csv(os.path.join(carpeta_salida, "varianza_pca_no_centrado_final.csv"), index=False)
    df_preprocesado.to_csv(os.path.join(carpeta_salida, "matriz_preprocesada_final.csv"), index=False)
    df_diagnostico_final.to_csv(os.path.join(carpeta_salida, "diagnostico_t2_q_no_centrado_final.csv"), index=False)
    df_eliminadas.to_csv(os.path.join(carpeta_salida, "muestras_eliminadas.csv"), index=False)

    print("\n[OK] Resultados exportados en:")
    print(carpeta_salida)
```

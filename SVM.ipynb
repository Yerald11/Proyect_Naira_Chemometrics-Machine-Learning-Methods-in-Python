```python
# ===========================================================================
# SVM / SVC INTERACTIVO - SOLO ENTRENAMIENTO / CALIBRACIÓN
# Flujo:
# CSV -> SNV -> Savitzky-Golay -> SVM preliminar
# -> diagnóstico -> eliminación manual -> SVM final
# ===========================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

plt.style.use("default")


# ===========================================================================
# FUNCIONES DE ENTRADA
# ===========================================================================

def pedir_texto(mensaje, default=None):
    if default is None:
        return input(mensaje).strip()
    txt = input(f"{mensaje} [{default}]: ").strip()
    return default if txt == "" else txt


def pedir_entero(mensaje, default=None, minimo=None):
    while True:
        txt = pedir_texto(mensaje, default)
        try:
            valor = int(txt)
            if minimo is not None and valor < minimo:
                print(f"Debe ser >= {minimo}")
                continue
            return valor
        except ValueError:
            print("Ingresa un número entero válido.")


def pedir_float(mensaje, default=None, minimo=None):
    while True:
        txt = pedir_texto(mensaje, default)
        try:
            valor = float(txt)
            if minimo is not None and valor < minimo:
                print(f"Debe ser >= {minimo}")
                continue
            return valor
        except ValueError:
            print("Ingresa un número válido.")


def pedir_si_no(mensaje, default="s"):
    while True:
        txt = pedir_texto(f"{mensaje} (s/n)", default).lower()
        if txt in ["s", "si", "sí", "y", "yes"]:
            return True
        if txt in ["n", "no"]:
            return False
        print("Responde con s o n.")


def pedir_kernel():
    while True:
        txt = pedir_texto("Kernel SVM: linear, rbf o poly", "rbf").lower()
        if txt in ["linear", "rbf", "poly"]:
            return txt
        print("Kernel inválido. Usa linear, rbf o poly.")


def pedir_gamma():
    while True:
        txt = pedir_texto("Gamma para SVM: scale, auto o número", "scale").lower()
        if txt in ["scale", "auto"]:
            return txt
        try:
            return float(txt)
        except ValueError:
            print("Gamma debe ser scale, auto o un número.")


# ===========================================================================
# PREPROCESAMIENTO
# ===========================================================================

def listar_csv(carpeta):
    archivos = [f for f in os.listdir(carpeta) if f.lower().endswith(".csv")]
    archivos.sort()
    return archivos


def aplicar_snv(X, ddof=0):
    media = X.mean(axis=1, keepdims=True)
    std = X.std(axis=1, ddof=ddof, keepdims=True)

    if np.any(std == 0):
        raise ValueError("Hay espectros con desviación estándar cero. No se puede aplicar SNV.")

    return (X - media) / std


def aplicar_savgol(X, longitudes_onda, window_length, polyorder, deriv):
    if window_length % 2 == 0:
        raise ValueError("La ventana Savitzky-Golay debe ser impar.")

    if window_length <= polyorder:
        raise ValueError("La ventana debe ser mayor que el orden del polinomio.")

    diffs = np.diff(longitudes_onda)

    if not np.allclose(diffs, diffs[0]):
        print("[ADVERTENCIA] Longitudes de onda no perfectamente equiespaciadas. Se usará delta promedio.")

    delta = np.mean(diffs)

    X_sg = savgol_filter(
        X,
        window_length=window_length,
        polyorder=polyorder,
        deriv=deriv,
        delta=delta,
        axis=1,
        mode="interp"
    )

    return X_sg, delta


# ===========================================================================
# SVM
# ===========================================================================

def entrenar_svm(X, y, kernel="rbf", C=1.0, gamma="scale", degree=3, usar_scaler=True):
    pasos = []

    if usar_scaler:
        pasos.append(("scaler", StandardScaler()))

    svm = SVC(
        kernel=kernel,
        C=C,
        gamma=gamma,
        degree=degree,
        probability=True,
        decision_function_shape="ovr"
    )

    pasos.append(("svm", svm))

    modelo = Pipeline(pasos)
    modelo.fit(X, y)

    y_pred = modelo.predict(X)
    decision = modelo.decision_function(X)

    try:
        prob = modelo.predict_proba(X)
    except Exception:
        prob = None

    return {
        "modelo": modelo,
        "y_pred": y_pred,
        "decision": decision,
        "prob": prob,
        "clases": modelo.named_steps["svm"].classes_
    }


def diagnostico_svm(X, svm_res):
    modelo = svm_res["modelo"]
    svm = modelo.named_steps["svm"]

    y_pred = svm_res["y_pred"]

    decision = svm_res["decision"]

    if decision.ndim == 1:
        margen = np.abs(decision)
    else:
        margen = np.max(decision, axis=1) - np.partition(decision, -2, axis=1)[:, -2]

    soporte_indices = svm.support_

    es_soporte = np.zeros(X.shape[0], dtype=bool)
    es_soporte[soporte_indices] = True

    return margen, es_soporte


def graficar_espectros(X_raw, X_proc, longitudes_onda):
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    ax[0].plot(longitudes_onda, X_raw.T, alpha=0.45)
    ax[0].set_title("Espectros originales")
    ax[0].set_xlabel("Longitud de onda")
    ax[0].set_ylabel("Absorbancia")
    ax[0].grid(alpha=0.2)

    ax[1].plot(longitudes_onda, X_proc.T, alpha=0.45)
    ax[1].set_title("Espectros preprocesados")
    ax[1].set_xlabel("Longitud de onda")
    ax[1].set_ylabel("Intensidad preprocesada")
    ax[1].grid(alpha=0.2)

    plt.tight_layout()
    plt.show()


def graficar_svm_diagnostico(X_proc, y_labels, svm_res, margen, es_soporte, titulo="preliminar"):
    pca = PCA(n_components=2)
    scores = pca.fit_transform(X_proc)

    fig, ax = plt.subplots(1, 3, figsize=(19, 5))

    for clase in np.unique(y_labels):
        idx = y_labels == clase
        ax[0].scatter(
            scores[idx, 0],
            scores[idx, 1],
            label=clase,
            s=65,
            edgecolors="black",
            linewidths=0.6,
            alpha=0.85
        )

    ax[0].scatter(
        scores[es_soporte, 0],
        scores[es_soporte, 1],
        facecolors="none",
        edgecolors="red",
        s=130,
        linewidths=1.5,
        label="Vectores soporte"
    )

    for i in range(scores.shape[0]):
        ax[0].text(scores[i, 0], scores[i, 1], str(i), fontsize=8, alpha=0.65)

    ax[0].set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}%)")
    ax[0].set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}%)")
    ax[0].set_title(f"PCA visual + vectores soporte - {titulo}")
    ax[0].legend()
    ax[0].grid(alpha=0.2)

    colores_margen = np.where(es_soporte, "crimson", "teal")
    ax[1].bar(range(len(margen)), margen, color=colores_margen, alpha=0.8)
    ax[1].set_title(f"Margen de decisión SVM - {titulo}")
    ax[1].set_xlabel("Índice")
    ax[1].set_ylabel("Margen")
    ax[1].grid(axis="y", alpha=0.25)

    pred = svm_res["y_pred"]
    correcto = pred == y_labels
    colores_error = np.where(correcto, "teal", "crimson")

    ax[2].bar(range(len(correcto)), correcto.astype(int), color=colores_error, alpha=0.8)
    ax[2].set_title(f"Clasificación aparente - {titulo}")
    ax[2].set_xlabel("Índice")
    ax[2].set_ylabel("Correcta = 1")
    ax[2].set_ylim(-0.05, 1.05)
    ax[2].grid(axis="y", alpha=0.25)

    plt.tight_layout()
    plt.show()


def graficar_probabilidades(svm_res, titulo="final"):
    prob = svm_res["prob"]
    clases = svm_res["clases"]

    if prob is None:
        return

    plt.figure(figsize=(11, 5))

    for j, clase in enumerate(clases):
        plt.plot(prob[:, j], marker="o", linestyle="", label=f"Probabilidad {clase}")

    plt.xlabel("Índice de muestra")
    plt.ylabel("Probabilidad estimada")
    plt.title(f"Probabilidades SVM - {titulo}")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()


# ===========================================================================
# 1. CONFIGURACIÓN INTERACTIVA
# ===========================================================================

print("=" * 75)
print("SVM / SVC INTERACTIVO - SOLO ENTRENAMIENTO / CALIBRACIÓN")
print("Flujo: SNV + Savitzky-Golay + SVM preliminar + descarte manual opcional")
print("=" * 75)

carpeta = pedir_texto("\nIngresa la carpeta donde están los CSV", ".")
carpeta = os.path.abspath(carpeta)

if not os.path.isdir(carpeta):
    raise FileNotFoundError(f"No existe la carpeta: {carpeta}")

csv_disponibles = listar_csv(carpeta)

if len(csv_disponibles) == 0:
    raise FileNotFoundError("No se encontraron archivos CSV.")

print("\nCSV encontrados:")
for i, archivo in enumerate(csv_disponibles, start=1):
    print(f"{i}. {archivo}")

seleccion_txt = pedir_texto("\nEscribe los números de los CSV separados por coma. Ejemplo: 1,2,3")
indices_csv = [int(x.strip()) - 1 for x in seleccion_txt.split(",")]
archivos_seleccionados = [csv_disponibles[i] for i in indices_csv]

etiquetas_clase = {}

print("\nAsigna una etiqueta/clase para cada archivo.")
for archivo in archivos_seleccionados:
    etiqueta_default = os.path.splitext(archivo)[0]
    etiqueta = pedir_texto(f"Etiqueta para {archivo}", etiqueta_default)
    etiquetas_clase[archivo] = etiqueta

sg_window = pedir_entero("\nVentana Savitzky-Golay", default=17, minimo=3)
sg_polyorder = pedir_entero("Orden del polinomio Savitzky-Golay", default=2, minimo=1)
sg_deriv = pedir_entero("Orden de derivada Savitzky-Golay", default=1, minimo=0)

kernel = pedir_kernel()
C = pedir_float("\nParámetro C del SVM", default=1.0, minimo=0.000001)
gamma = pedir_gamma()

degree = 3
if kernel == "poly":
    degree = pedir_entero("Grado del kernel polinomial", default=3, minimo=2)

usar_scaler = pedir_si_no("\n¿Aplicar autoscaling por columna antes del SVM?", default="s")
exportar = pedir_si_no("\n¿Deseas exportar resultados CSV?", default="s")


# ===========================================================================
# 2. CARGA DE DATOS
# ===========================================================================

dfs = []
y_labels = []

for archivo in archivos_seleccionados:
    ruta = os.path.join(carpeta, archivo)
    df = pd.read_csv(ruta)
    clase = etiquetas_clase[archivo]

    dfs.append(df)
    y_labels.extend([clase] * len(df))

    print(f"[OK] {archivo}: {df.shape[0]} muestras, clase = {clase}")

df_global = pd.concat(dfs, ignore_index=True)
y_labels = np.array(y_labels)

columnas_convertidas = pd.to_numeric(df_global.columns, errors="coerce")
columnas_espectrales = df_global.columns[columnas_convertidas.notna()]
columnas_metadata = df_global.columns[columnas_convertidas.isna()]

longitudes_onda = columnas_convertidas[columnas_convertidas.notna()].astype(float).values
X_raw = df_global[columnas_espectrales].astype(float).values

orden = np.argsort(longitudes_onda)
longitudes_onda = longitudes_onda[orden]
X_raw = X_raw[:, orden]

print("\nColumnas de metadatos omitidas:")
print(list(columnas_metadata))

print(f"\nMatriz espectral cruda: {X_raw.shape} = muestras x variables")

if np.isnan(X_raw).any():
    raise ValueError("La matriz contiene NaN.")

if np.isinf(X_raw).any():
    raise ValueError("La matriz contiene infinitos.")


# ===========================================================================
# 3. PREPROCESAMIENTO
# ===========================================================================

X_snv = aplicar_snv(X_raw, ddof=0)

X_proc, delta_sg = aplicar_savgol(
    X_snv,
    longitudes_onda,
    window_length=sg_window,
    polyorder=sg_polyorder,
    deriv=sg_deriv
)

print("\nPreprocesamiento aplicado:")
print("- SNV muestra por muestra, ddof=0")
print(f"- Savitzky-Golay: ventana={sg_window}, polinomio={sg_polyorder}, derivada={sg_deriv}, delta={delta_sg:.6f}")


# ===========================================================================
# 4. SVM PRELIMINAR
# ===========================================================================

svm_pre = entrenar_svm(
    X_proc,
    y_labels,
    kernel=kernel,
    C=C,
    gamma=gamma,
    degree=degree,
    usar_scaler=usar_scaler
)

margen_pre, es_soporte_pre = diagnostico_svm(X_proc, svm_pre)

print("\n" + "=" * 75)
print("SVM PRELIMINAR - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels, svm_pre["y_pred"]))

graficar_espectros(X_raw, X_proc, longitudes_onda)

graficar_svm_diagnostico(
    X_proc,
    y_labels,
    svm_pre,
    margen_pre,
    es_soporte_pre,
    titulo="preliminar"
)

graficar_probabilidades(svm_pre, titulo="preliminar")

cm = confusion_matrix(y_labels, svm_pre["y_pred"], labels=svm_pre["clases"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=svm_pre["clases"])
disp.plot(cmap="Blues", values_format="d")
plt.title("Matriz de confusión aparente - SVM preliminar")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

df_diag_pre = pd.DataFrame({
    "Indice": np.arange(len(y_labels)),
    "Clase_real": y_labels,
    "Clase_predicha": svm_pre["y_pred"],
    "Correcta": y_labels == svm_pre["y_pred"],
    "Margen_decision": margen_pre,
    "Es_vector_soporte": es_soporte_pre
})

if svm_pre["prob"] is not None:
    for j, clase in enumerate(svm_pre["clases"]):
        df_diag_pre[f"Probabilidad_{clase}"] = svm_pre["prob"][:, j]

print("\nMuestras con menor margen de decisión:")
display(df_diag_pre.sort_values("Margen_decision", ascending=True).head(15))

print("\nVectores soporte:")
display(df_diag_pre[df_diag_pre["Es_vector_soporte"] == True])

print("\nMuestras mal clasificadas en calibración aparente:")
display(df_diag_pre[df_diag_pre["Correcta"] == False])


# ===========================================================================
# 5. ELIMINACIÓN MANUAL DESPUÉS DEL DIAGNÓSTICO
# ===========================================================================

eliminar = pedir_si_no(
    "\nDespués de ver los diagnósticos SVM, ¿deseas eliminar muestras manualmente?",
    default="n"
)

indices_eliminar = []

if eliminar:
    txt = pedir_texto("Escribe los índices globales a eliminar separados por coma. Ejemplo: 54 o 54,67")
    if txt.strip() != "":
        indices_eliminar = [int(x.strip()) for x in txt.split(",")]

if len(indices_eliminar) > 0:
    indices_eliminar = np.array(indices_eliminar, dtype=int)

    if np.any(indices_eliminar < 0) or np.any(indices_eliminar >= X_raw.shape[0]):
        raise ValueError("Hay índices a eliminar fuera del rango.")

    mascara = np.ones(X_raw.shape[0], dtype=bool)
    mascara[indices_eliminar] = False

    print("\nMuestras eliminadas manualmente:")
    for idx in indices_eliminar:
        print(f"- Índice global {idx}, clase {y_labels[idx]}")

    X_raw_final = X_raw[mascara]
    X_proc_final = X_proc[mascara]
    y_labels_final = y_labels[mascara]
    df_global_final = df_global.loc[mascara].reset_index(drop=True)

else:
    print("\nNo se eliminaron muestras. El SVM final usará todas las muestras.")
    X_raw_final = X_raw.copy()
    X_proc_final = X_proc.copy()
    y_labels_final = y_labels.copy()
    df_global_final = df_global.copy()

print(f"\nMatriz final para SVM: {X_proc_final.shape}")


# ===========================================================================
# 6. SVM FINAL
# ===========================================================================

svm_final = entrenar_svm(
    X_proc_final,
    y_labels_final,
    kernel=kernel,
    C=C,
    gamma=gamma,
    degree=degree,
    usar_scaler=usar_scaler
)

margen_final, es_soporte_final = diagnostico_svm(X_proc_final, svm_final)

print("\n" + "=" * 75)
print("SVM FINAL - CALIBRACIÓN")
print("=" * 75)

print(classification_report(y_labels_final, svm_final["y_pred"]))

graficar_espectros(X_raw_final, X_proc_final, longitudes_onda)

graficar_svm_diagnostico(
    X_proc_final,
    y_labels_final,
    svm_final,
    margen_final,
    es_soporte_final,
    titulo="final"
)

graficar_probabilidades(svm_final, titulo="final")

cm = confusion_matrix(y_labels_final, svm_final["y_pred"], labels=svm_final["clases"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=svm_final["clases"])
disp.plot(cmap="Blues", values_format="d")
plt.title("Matriz de confusión aparente - SVM final")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ===========================================================================
# 7. TABLAS FINALES
# ===========================================================================

df_pred_final = pd.DataFrame({
    "Indice": np.arange(len(y_labels_final)),
    "Clase_real": y_labels_final,
    "Clase_predicha_calibracion": svm_final["y_pred"],
    "Correcta": y_labels_final == svm_final["y_pred"],
    "Margen_decision": margen_final,
    "Es_vector_soporte": es_soporte_final
})

if svm_final["prob"] is not None:
    for j, clase in enumerate(svm_final["clases"]):
        df_pred_final[f"Probabilidad_{clase}"] = svm_final["prob"][:, j]

display(df_pred_final)

print("\nMuestras con menor margen final:")
display(df_pred_final.sort_values("Margen_decision", ascending=True).head(15))

print("\nVectores soporte finales:")
display(df_pred_final[df_pred_final["Es_vector_soporte"] == True])


# ===========================================================================
# 8. EXPORTACIÓN
# ===========================================================================

if exportar:
    carpeta_salida = os.path.join(carpeta, "resultados_svm_entrenamiento")
    os.makedirs(carpeta_salida, exist_ok=True)

    df_preprocesado = pd.DataFrame(X_proc_final, columns=[str(w) for w in longitudes_onda])
    df_preprocesado.insert(0, "Clase", y_labels_final)
    df_preprocesado.insert(0, "Indice", np.arange(len(y_labels_final)))

    df_eliminadas = pd.DataFrame({
        "Indice_original_eliminado": indices_eliminar
    })

    df_pred_final.to_csv(os.path.join(carpeta_salida, "svm_predicciones_calibracion_final.csv"), index=False)
    df_diag_pre.to_csv(os.path.join(carpeta_salida, "svm_diagnostico_preliminar.csv"), index=False)
    df_preprocesado.to_csv(os.path.join(carpeta_salida, "matriz_preprocesada_snv_sg_final.csv"), index=False)
    df_eliminadas.to_csv(os.path.join(carpeta_salida, "muestras_eliminadas.csv"), index=False)

    print("\n[OK] Resultados exportados en:")
    print(carpeta_salida)
```
